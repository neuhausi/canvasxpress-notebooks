# Enhancing Analysis with Faceted Charts

Facets allow for easy comparison of data subsets through multiple plots within a single chart, enhancing analysis and storytelling.

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/layout-15.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).
