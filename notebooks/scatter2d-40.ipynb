# Calibration curve (model evaluation)

Reliability curve from CanvasXpress.calibrationConfig with the perfect-calibration diagonal and ECE.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/scatter2d-40.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "scatter2d40",
    data = {
      "y": {
        "vars": ["b0", "b1", "b2", "b3", "b4", "b5", "b6", "b7", "b8", "b9"],
        "smps": ["Predicted", "Observed"],
        "data": [
          [0.05506043758534344, 0.08888888888888889],
          [0.15350885244491658, 0.1],
          [0.24588810780438444, 0.21739130434782608],
          [0.3483882165938756, 0.42592592592592593],
          [0.44956108423491237, 0.45652173913043476],
          [0.5571396615214862, 0.5517241379310345],
          [0.6415745994914845, 0.6071428571428571],
          [0.7567069294784358, 0.7321428571428571],
          [0.8505518800465788, 0.7391304347826086],
          [0.9524477263592037, 0.9302325581395349]
        ]
      }
    },
    config = {
      "graphType": "Scatter2D",
      "scatterType": "line",
      "xAxis": ["Predicted"],
      "yAxis": ["Observed"],
      "setMinX": 0,
      "setMaxX": 1,
      "setMinY": 0,
      "setMaxY": 1,
      "xAxisTitle": "Mean predicted probability",
      "yAxisTitle": "Observed frequency",
      "title": "Calibration — probability estimates",
      "showLoessFit": False,
      "showTransition": False,
      "decorations": {
        "line": [
          {
            "x": 0,
            "y": 0,
            "x2": 1,
            "y2": 1,
            "color": "#999999",
            "width": 1,
            "type": "dashed",
            "label": "Perfect"
          }
        ]
      },
      "colors": ["#2f6df0"]
    },
    width = 600,
    height = 600
  )
)
```
