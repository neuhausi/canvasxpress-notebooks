# Two diverging series on dual axes

Atmospheric CO2 rising on the left axis and sea-ice extent falling on the right axis over the same years, each line in its own colour with the axis labels matched to the series. AreaLine with xAxis and xAxis2.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/time-series-18.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).
