# Mastering Multi-Series Bar Graphs

Explore a basic bar graph with multiple series, showcasing variations in data effectively.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/bar-3.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "bar3",
    data = {
      "y": {
        "data": [
          [4, 5, 4, 4, 7],
          [3, 6, 5, 5, 6],
          [1, 1, 1, 3, 5]
        ],
        "smps": ["Cat 1", "Cat 2", "Cat 3", "Cat 4", "Cat 5"],
        "vars": ["Var 1", "Var 2", "Var 3"]
      }
    },
    config = {
      "graphOrientation": "vertical",
      "graphType": "Bar",
      "legendColumns": 3,
      "legendPosition": "bottom",
      "smpTextRotate": 90,
      "smpTitle": "Categories",
      "title": "Bar graph with multiple series",
      "xAxis": ["Var 1", "Var 2", "Var 3"],
      "xAxisTitle": "Value"
    },
    width = 600,
    height = 600
  )
)
```
