# Efficient Project Planning with Gantt Chart

Visualize project timelines and dependencies effectively with a simple Gantt chart. 

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/gantt-1.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "gantt1",
    data = {
      "y": {
        "data": [
          [20140221, 20140521, 20140821, 20141121, 20150221, 20150521],
          [20140520, 20140820, 20141120, 20150220, 20150520, 20150820]
        ],
        "smps": ["S01", "S02", "S03", "S04", "S05", "S06"],
        "vars": ["Start", "End"]
      }
    },
    config = {
      "blockContrast": True,
      "ganttEnd": "End",
      "ganttStart": "Start",
      "graphType": "Gantt",
      "xAxis": ["Start", "End"]
    },
    width = 600,
    height = 600
  )
)
```
