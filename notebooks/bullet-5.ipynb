# Visualization with stylish Bullet Graphs

Clearly visualize performance with bullet graphs, incorporating a stylish bullet graphs and a custumized target marker.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/bullet-5.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "bullet5",
    data = {
      "y": {
        "data": [
          [250.43, 5.69, 9.79],
          [250, 3, 8],
          [400, 6, 10],
          [600, 12, 11],
          [520, 7.5, 10.5]
        ],
        "smps": ["Total Distance", "Average Distance", "Average Speed"],
        "vars": ["Value", "Bad", "Acceptable", "Good", "Target"]
      }
    },
    config = {
      "barType": "bullet",
      "rangeColors": ["#FFFFFF", "#F2F2F2", "#D8D8D8"],
      "rangeStack": ["Bad", "Acceptable", "Good"],
      "rangeStackShow": True,
      "bulletTargetType": "openCircle",
      "bulletTargetVarName": "Target",
      "bulletTargetWidthRatio": 0.2,
      "bulletStyle": "roundGradient",
      "bulletWidthRatio": 0.5,
      "colors": ["#9422F5"],
      "dataValuesPosition": "inside",
      "dataTextColor": "#FFFFFF",
      "dataTextScaleFontFactor": 0.8,
      "graphOrientation": "horizontal",
      "graphType": "Bullet",
      "layoutAdjust": True,
      "layoutSpacing": 20,
      "layoutTopology": "3X1",
      "segregateSamplesBy": ["sample"],
      "showDataValues": True,
      "showLegend": False,
      "showSampleNames": False,
      "stripBackgroundColor": "#FFFFFF",
      "stripTextAlign": "left",
      "stripTextColor": "#000000",
      "xAxis": ["Value"],
      "xAxis2Show": True,
      "xAxisGridMajorShow": False,
      "xAxisGridMinorShow": False,
      "xAxisShow": False
    },
    width = 600,
    height = 600
  )
)
```
