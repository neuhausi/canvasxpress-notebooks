# Advanced Data Analysis with 4-Parameter Equations

Explore the power of 4-parameter equation fitting for complex data analysis and visualization.

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/nonlinear-fit-1.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "nonlinearfit1",
    data = {
      "y": {
        "data": [
          [0.0009, 172],
          [0.0018, 177],
          [0.0037, 160],
          [0.0073, 166],
          [0.0146, 211],
          [0.0293, 248],
          [0.0586, 269],
          [0.117, 283],
          [0.234, 298],
          [0.469, 314],
          [0.938, 328],
          [1.88, 316]
        ],
        "smps": ["Concentration", "V1"],
        "vars": ["S1", "S2", "S3", "S4", "S5", "S6", "S7", "S8", "S9", "S10", "S11", "S12"]
      }
    },
    config = {
      "decorations": {
        "nlfit": [
          {
            "label": "Custom Fit",
            "param": [164, 313, 0.031, -1.5, 1.2e-06, 1.9],
            "type": "cst"
          },
          {
            "label": "Regular Fit",
            "param": [164, 313, 0.031, 1.5, 1.2e-06, 1.9],
            "type": "reg"
          }
        ]
      },
      "graphType": "Scatter2D",
      "setMaxY": 350,
      "setMinY": 100,
      "showDecorations": True,
      "xAxis": ["Concentration"],
      "xAxisTransform": "log10",
      "xAxisTransformTicks": True,
      "yAxisExact": True,
      "yAxis": ["V1"]
    },
    width = 600,
    height = 600
  )
)
```
