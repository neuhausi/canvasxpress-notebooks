# Clustered Heatmap with Multiple Annotation Tracks

The violin tracks are a new overlay type that reuses the same smpOverlays/varOverlays + *OverlayProperties mechanism as the Pie, Bar and Default tracks — set type:"Violin".

Live example, data and code in other languages: https://www.canvasxpress.org/examples/heatmap-20.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).
