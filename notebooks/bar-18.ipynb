# Bar Graph with Locale Number Formats

Live example, data and code in other languages: https://www.canvasxpress.org/examples/bar-18.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "bar18",
    data = {
      "y": {
        "vars": ["Revenue"],
        "smps": ["North America", "Europe", "Asia Pacific", "Latin America"],
        "data": [
          [12500000, 9400000, 15800000, 3100000]
        ]
      }
    },
    config = {
      "graphType": "Bar",
      "graphOrientation": "vertical",
      "title": "Annual revenue by region",
      "numberFormat": {
        "style": "currency",
        "currency": "USD",
        "decimals": 0
      },
      "dataLabelNumberFormat": "compact",
      "showDataValues": True,
      "legendShow": False,
      "citation": "numberFormat: currency axis, compact data labels"
    },
    width = 600,
    height = 600
  )
)
```
