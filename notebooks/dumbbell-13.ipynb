# Timespan ranges with dots and direct labels

Ten diseases as horizontal spans from the year the pathogen was identified to the year a vaccine was licensed, with a dot at each end and the span length as a direct label. Dumbbell graph on a year axis.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/dumbbell-13.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "dumbbell13",
    data = {
      "y": {
        "vars": ["Pathogen identified", "Vaccine licensed"],
        "smps": ["Cholera", "Tuberculosis", "Malaria", "Dengue", "Anthrax", "Rabies", "Hepatitis A", "Mumps", "HPV", "Ebola"],
        "data": [
          [1883, 1882, 1880, 1943, 1876, 1885, 1973, 1934, 1983, 1976],
          [1991, 1921, 2021, 2019, 1881, 1885, 1995, 1967, 2006, 2019]
        ]
      }
    },
    config = {
      "graphType": "Dumbbell",
      "dumbbellType": "connected",
      "theme": "paulTol",
      "colorScheme": "User",
      "colors": ["#332288", "#332288"],
      "title": "From pathogen to vaccine",
      "subtitle": "Year the cause of ten diseases was identified and the year a vaccine against it was licensed",
      "xAxis": ["Pathogen identified", "Vaccine licensed"],
      "xAxisTitle": "Year",
      "setMinX": 1860,
      "setMaxX": 2040,
      "xAxisTickFormat": "%.0f",
      "sortDir": "ascending",
      "showLegend": False,
      "dataPointSizeScaleFactor": 2.2,
      "citation": "Source: synthetic data for illustration"
    },
    width = 760,
    height = 480
  )
)
```
