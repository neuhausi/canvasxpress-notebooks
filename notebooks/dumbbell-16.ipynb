# Arrow plot coloured by size of change

Fifteen research fields as arrows from the 2014 to the 2024 open-access share, arrows coloured by the size of the change through an annotation, with a shaded majority region. Dumbbell graph with arrows and colorBy.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/dumbbell-16.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "dumbbell16",
    data = {
      "y": {
        "vars": ["2014", "2024"],
        "smps": ["Immunology", "Neuroscience", "Pharmacology", "Biochemistry", "Agronomy", "Psychology", "Economics", "Chemistry", "Materials", "Physics", "Computer science", "Mathematics", "Nursing", "Medicine", "Dentistry"],
        "data": [
          [40, 38, 39, 39, 33, 43, 18, 26, 25, 19, 16, 16, 36, 35, 28],
          [46.8, 47.8, 47.2, 45.8, 42.6, 57.3, 30, 32.3, 30.6, 23.7, 21, 19.9, 47.5, 45.6, 38.8]
        ]
      },
      "x": {
        "Change": ["5 to 10 points", "5 to 10 points", "5 to 10 points", "5 to 10 points", "5 to 10 points", "More than 10 points", "More than 10 points", "5 to 10 points", "5 to 10 points", "5 points or less", "5 points or less", "5 points or less", "More than 10 points", "More than 10 points", "More than 10 points"],
        "Domain": ["1. Life sciences", "1. Life sciences", "1. Life sciences", "1. Life sciences", "1. Life sciences", "2. Social sciences", "2. Social sciences", "3. Physical sciences", "3. Physical sciences", "3. Physical sciences", "3. Physical sciences", "3. Physical sciences", "4. Health", "4. Health", "4. Health"]
      }
    },
    config = {
      "graphType": "Dumbbell",
      "dumbbellType": "arrow",
      "dumbbellArrowDirection": "order",
      "theme": "classic",
      "colorScheme": "User",
      "colors": ["#6b6b6b", "#009E73"],
      "title": "Open-access share of papers by field",
      "subtitle": "Share of papers published open access, grey 2014 to green 2024 (%); shaded: majority open access",
      "xAxis": ["2014", "2024"],
      "xAxisTitle": "Share open access (%)",
      "connectByColor": "#009E73",
      "connectByWidth": 3,
      "dataPointSizeScaleFactor": 0.7,
      "setMinX": 0,
      "setMaxX": 70,
      "showLegend": False,
      "smpLabelInterval": 1,
      "segregateSamplesBy": ["Domain"],
      "stripBackgroundColor": "rgba(255,255,255,0)",
      "stripTextAlign": "left",
      "stripBackgroundBorderColor": "rgba(255,255,255,0)",
      "layoutTopology": "4X1",
      "decorations": {
        "range": [
          {
            "value": 50,
            "value2": 70,
            "color": "rgba(0,158,115,0.10)",
            "tooltip": "Majority open access"
          }
        ],
        "text": [
          {
            "value": 38,
            "smpId": 4,
            "panel": 3,
            "label": "Over 50% open access ▶",
            "color": "#009E73",
            "labelScaleFontFactor": 0.8
          }
        ]
      },
      "citation": "Source: synthetic data for illustration"
    },
    width = 760,
    height = 760
  )
)
```
