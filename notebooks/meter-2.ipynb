# Speedometer Graph: Visual Performance Metrics Monitoring

A speedometer-style meter graph displays data in a visual, intuitive format ideal for monitoring and analyzing performance metrics.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/meter-2.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "meter2",
    data = {
      "y": {
        "data": [
          [85]
        ],
        "smps": ["January"],
        "vars": ["Performance"]
      }
    },
    config = {
      "graphType": "Meter",
      "setMax": 200,
      "setMin": 0,
      "rangeSegments": [0, 25, 50, 75, 200],
      "meterType": "speedometer",
      "xAxis": ["Performance"]
    },
    width = 600,
    height = 600
  )
)
```
