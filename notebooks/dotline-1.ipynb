# DotLine Graphs: Visualizing Data Distribution

Visualize data distribution effectively with DotLine graphs, showing each data point as a dot above a number line for easy comparison.

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/dotline-1.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "dotline1",
    data = {
      "x": {
        "Factor1": ["Lev : 1", "Lev : 2", "Lev : 3", "Lev : 1", "Lev : 2", "Lev : 3"],
        "Factor2": ["Lev : A", "Lev : B", "Lev : A", "Lev : B", "Lev : A", "Lev : B"],
        "Factor3": ["Lev : X", "Lev : X", "Lev : Y", "Lev : Y", "Lev : Z", "Lev : Z"],
        "Factor4": [5, 10, 15, 20, 25, 30],
        "Factor5": [8, 16, 24, 32, 40, 48],
        "Factor6": [10, 20, 30, 40, 50, 60]
      },
      "y": {
        "data": [
          [5, 10, 25, 40, 45, 50],
          [95, 80, 75, 70, 55, 40],
          [25, 30, 45, 60, 65, 70],
          [55, 40, 35, 30, 15, 1]
        ],
        "desc": ["Magnitude1", "Magnitude2"],
        "smps": ["S1", "S2", "S3", "S4", "S5", "S6"],
        "vars": ["V1", "V2", "V3", "V4"]
      },
      "z": {
        "Annt1": ["Desc : 1", "Desc : 2", "Desc : 3", "Desc : 4"],
        "Annt2": ["Desc : A", "Desc : B", "Desc : A", "Desc : B"],
        "Annt3": ["Desc : X", "Desc : X", "Desc : Y", "Desc : Y"],
        "Annt4": [5, 10, 15, 20],
        "Annt5": [8, 16, 24, 32],
        "Annt6": [10, 20, 30, 40]
      }
    },
    config = {
      "graphOrientation": "vertical",
      "graphType": "DotLine",
      "legendColumns": 2,
      "legendPosition": "bottom",
      "lineThickness": 2,
      "lineType": "spline",
      "smpTextRotate": 45,
      "smpTitle": "Collection of Samples",
      "smpTitleFontStyle": "italic",
      "subtitle": "Random Data",
      "title": "Dot-Line Graphs",
      "xAxis": ["V1", "V2"],
      "xAxis2": ["V3", "V4"],
      "xAxisGridMajorColor": "rgb(0,0,0)"
    },
    width = 600,
    height = 600
  )
)
```
