# Editorial Theme with Responsive Rules

Live example, data and code in other languages: https://www.canvasxpress.org/examples/area-17.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "area17",
    data = {
      "y": {
        "vars": ["Coal", "Gas", "Wind", "Solar"],
        "smps": ["Q1", "Q2", "Q3", "Q4", "Q5", "Q6", "Q7", "Q8"],
        "data": [
          [30, 28, 26, 24, 20, 18, 16, 14],
          [24, 25, 25, 26, 24, 23, 22, 21],
          [8, 9, 11, 13, 15, 17, 19, 21],
          [2, 3, 4, 5, 7, 9, 11, 13]
        ]
      }
    },
    config = {
      "graphType": "Area",
      "areaType": "stacked",
      "theme": "editorial",
      "title": "Generation share",
      "subtitle": "Percent of total, last eight quarters",
      "legendPosition": "right",
      "source": "IEA",
      "byline": "Chart: CanvasXpress",
      "responsiveRules": [
        {
          "maxWidth": 480,
          "config": {
            "legendPosition": "bottom",
            "fontScaleFontFactor": 0.85
          }
        }
      ],
      "graphOrientation": "vertical"
    },
    width = 600,
    height = 600
  )
)
```
