# Visualize Data with Spider Plots

Spider plots, also known as radar charts, display multivariate data in a single graphic, making them ideal for visualizing and comparing multiple variables at once.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/scatter2d-2.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).
