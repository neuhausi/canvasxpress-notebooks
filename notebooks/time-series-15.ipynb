# Monthly grouped columns: an outbreak and its decline

Thirty months of clinic cases by cause as grouped columns, one series dominating during the epidemic spike and decaying after an intervention, with a vertical marker at the intervention. Bar graph with grouped monthly samples.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/time-series-15.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "timeseries15",
    data = {
      "y": {
        "vars": ["Cholera", "Typhoid", "Other"],
        "smps": ["Apr. 1854", "May 1854", "June 1854", "July 1854", "Aug. 1854", "Sept. 1854", "Oct. 1854", "Nov. 1854", "Dec. 1854", "Jan. 1855", "Feb. 1855", "Mar. 1855", "Apr. 1855", "May 1855", "June 1855", "July 1855", "Aug. 1855", "Sept. 1855", "Oct. 1855", "Nov. 1855", "Dec. 1855", "Jan. 1856", "Feb. 1856", "Mar. 1856", "Apr. 1856", "May 1856", "June 1856", "July 1856", "Aug. 1856", "Sept. 1856"],
        "data": [
          [3, 4, 5, 6, 110, 234, 359, 459, 540, 704, 772, 445, 352, 236, 159, 33, 31, 15, 17, 12, 12, 10, 6, 6, 8, 6, 7, 7, 7, 5],
          [25, 14, 27, 8, 17, 26, 56, 72, 32, 74, 74, 71, 25, 60, 5, 25, 18, 33, 8, 13, 11, 11, 17, 9, 19, 20, 20, 33, 7, 31],
          [14, 22, 28, 28, 12, 22, 22, 14, 98, 33, 28, 56, 25, 34, 23, 12, 23, 20, 12, 10, 24, 24, 13, 25, 17, 9, 15, 20, 27, 27]
        ]
      }
    },
    config = {
      "graphType": "Bar",
      "titleAlign": "plot::0",
      "subtitleAlign": "plot::0",
      "legendHorizontalJustification": 0,
      "legendJustifyTo": "plot",
      "graphOrientation": "vertical",
      "theme": "ggplot",
      "colorScheme": "User",
      "colors": ["#56B4E9", "#E1785B", "#8A8A8A"],
      "xAxis": ["Cholera", "Typhoid", "Other"],
      "title": "A cholera outbreak and the effect of a clean-water intervention",
      "subtitle": "Clinic cases per month in a fictional river town, by cause. Illustrative data.",
      "smpLabelInterval": 6,
      "smpTextRotate": 90,
      "showLegend": True,
      "legendPosition": "top",
      "xAxisTitle": "Cases per month",
      "smpTitle": "",
      "decorations": {
        "line": [
          {
            "value": 45,
            "color": "#B5503C",
            "width": 2.5,
            "type": "dotted",
            "label": "Pre-outbreak clinic baseline",
            "align": "right"
          }
        ],
        "marker": [
          {
            "text": "Clean-water\nintervention begins",
            "variable": "Cholera",
            "sample": "Aug. 1855",
            "type": "curved",
            "x": 0.5,
            "y": 0.45
          }
        ]
      }
    },
    width = 760,
    height = 560
  )
)
```
