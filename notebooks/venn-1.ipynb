# 4-Way Venn Diagram for Relationship Visualization

Visualize overlapping relationships between four sets using a Venn diagram.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/venn-1.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "venn1",
    data = {
      "venn": {
        "data": {
          "A": 340,
          "AB": 639,
          "ABC": 552,
          "ABCD": 148,
          "ABD": 578,
          "AC": 456,
          "ACD": 298,
          "AD": 257,
          "B": 562,
          "BC": 915,
          "BCD": 613,
          "BD": 354,
          "C": 620,
          "CD": 143,
          "D": 592
        },
        "legend": {
          "A": "List 1",
          "B": "List 2",
          "C": "List 3",
          "D": "List 4"
        }
      }
    },
    config = {
      "graphType": "Venn",
      "showTransition": False,
      "vennGroups": 4
    },
    width = 600,
    height = 600
  )
)
```
