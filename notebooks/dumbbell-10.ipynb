# Three dots per row on a connecting range

Twelve species, each a row with minimum, optimum and maximum values as three coloured dots joined by a light connecting bar, rows sorted by one value, legend on top and the value axis at the top. Dumbbell graph with three series.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/dumbbell-10.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "dumbbell10",
    data = {
      "y": {
        "vars": ["Minimum (CTmin)", "Optimum", "Maximum (CTmax)"],
        "smps": ["Clownfish", "Blue tang", "Parrotfish", "Moray eel", "Lionfish", "Damselfish", "Butterflyfish", "Grouper", "Wrasse", "Goby", "Seahorse", "Pufferfish"],
        "data": [
          [19.5, 18.2, 17, 15.8, 14.9, 16.4, 18.8, 14.2, 13.5, 12.1, 15, 16.2],
          [27.5, 26, 25.1, 24, 23.2, 25.5, 26.8, 22.4, 21.5, 20, 22.8, 24.4],
          [34, 33.1, 33.4, 32.2, 31, 32.8, 33.6, 31.2, 30, 29.4, 29.8, 32]
        ]
      }
    },
    config = {
      "graphType": "Dumbbell",
      "dumbbellType": "connected",
      "theme": "classic",
      "colorScheme": "User",
      "colors": ["#56B4E9", "#555555", "#E69F00"],
      "title": "Reef fish tolerate a narrower warm range than a cold one",
      "subtitle": "Thermal limits and optimum of twelve reef fish species, in °C",
      "xAxisTitle": "Temperature (°C)",
      "sortDir": "ascending",
      "sortBy": "Maximum (CTmax)",
      "xAxis": ["Minimum (CTmin)", "Optimum", "Maximum (CTmax)"],
      "legendPosition": "top",
      "dataPointSizeScaleFactor": 2.2,
      "smpTitleFontStyle": "bold",
      "citation": "Source: synthetic data for illustration"
    },
    width = 760,
    height = 520
  )
)
```
