# Unlock Data Insights with Logarithmic Fit

Discover the power of logarithmic fit for accurately analyzing and visualizing complex data relationships.

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/nonlinear-fit-3.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "nonlinearfit3",
    data = {
      "y": {
        "data": [
          [1, 6],
          [2, 9.5],
          [3, 13],
          [4, 15],
          [5, 16.5],
          [6, 17.5],
          [7, 18.5],
          [8, 19],
          [9, 19.5],
          [10, 19.7],
          [11, 19.8]
        ],
        "smps": ["Age", "Height"],
        "vars": ["S1", "S2", "S3", "S4", "S5", "S6", "S7", "S8", "S9", "S10", "S11"]
      }
    },
    config = {
      "decorations": {
        "logarithmic": [
          {
            "color": "red"
          }
        ]
      },
      "graphType": "Scatter2D",
      "showDecorations": True,
      "showLegend": False,
      "xAxis": ["Age"],
      "yAxis": ["Height"]
    },
    width = 600,
    height = 600
  )
)
```
