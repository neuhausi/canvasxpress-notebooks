# US states choropleth with Alaska and Hawaii insets

A state-level rate on the Albers projection with Alaska and Hawaii as insets, state-name and value labels inside each state and a continuous viridis-like ramp with end labels. Map with mapAlbersInsets and showMapLabels.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/map-20.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).
