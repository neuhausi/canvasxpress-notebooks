# State Meter Card

A traffic-light state meter framed as a dashboard card, sized with meterThickness to always fit the card.

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/meter-14.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "meter14",
    data = {
      "y": {
        "data": [
          [85]
        ],
        "smps": ["January"],
        "vars": ["Performance"]
      }
    },
    config = {
      "graphType": "Meter",
      "setMax": 100,
      "setMin": 0,
      "meterType": "state",
      "meterCard": True,
      "meterThickness": 0.8,
      "xAxis": ["Performance"]
    },
    width = 600,
    height = 420
  )
)
```
