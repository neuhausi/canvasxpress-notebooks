# Feature importance (model evaluation)

Horizontal feature-importance bar chart from CanvasXpress.importanceConfig, sorted by importance.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/bar-24.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "bar24",
    data = {
      "y": {
        "vars": ["Importance"],
        "smps": ["Glucose", "BMI", "Age", "BloodPressure", "Insulin", "Pedigree", "SkinThickness", "Pregnancies"],
        "data": [
          [0.28, 0.22, 0.17, 0.11, 0.09, 0.07, 0.04, 0.02]
        ]
      }
    },
    config = {
      "graphType": "Bar",
      "graphOrientation": "horizontal",
      "title": "Feature importance — diabetes model",
      "smpTitle": "Feature",
      "xAxisTitle": "Importance",
      "showLegend": False,
      "showTransition": False
    },
    width = 600,
    height = 600
  )
)
```
