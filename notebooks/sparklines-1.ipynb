# Sparklines — trend strips with reference line and endpoint markers

Sparklines show one compact trend strip per series with free per-panel y-scales, so the shape of each series reads at a glance. Here monthly sessions for four channels, each with a dashed mean reference line and endpoint markers.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/sparklines-1.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "sparklines1",
    data = {
      "y": {
        "vars": ["Sessions"],
        "smps": ["Jan-A", "Feb-A", "Mar-A", "Apr-A", "May-A", "Jun-A", "Jan-B", "Feb-B", "Mar-B", "Apr-B", "May-B", "Jun-B", "Jan-C", "Feb-C", "Mar-C", "Apr-C", "May-C", "Jun-C", "Jan-D", "Feb-D", "Mar-D", "Apr-D", "May-D", "Jun-D"],
        "data": [
          [120, 135, 128, 150, 162, 158, 60, 58, 64, 70, 69, 74, 210, 205, 230, 225, 248, 260, 44, 52, 49, 58, 55, 63]
        ]
      },
      "x": {
        "Channel": ["Organic", "Organic", "Organic", "Organic", "Organic", "Organic", "Email", "Email", "Email", "Email", "Email", "Email", "Paid", "Paid", "Paid", "Paid", "Paid", "Paid", "Referral", "Referral", "Referral", "Referral", "Referral", "Referral"]
      }
    },
    config = {
      "graphType": "Sparklines",
      "sparklineType": "line",
      "sparklineBaseline": "mean",
      "sparklineMarkers": "endpoints",
      "title": "Monthly sessions by channel"
    },
    width = 600,
    height = 600
  )
)
```
