# Arrow Plot (Before to After)

Live example, data and code in other languages: https://www.canvasxpress.org/examples/dumbbell-9.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "dumbbell9",
    data = {
      "y": {
        "vars": ["2014", "2024"],
        "smps": ["USA", "Germany", "Japan", "Brazil", "India", "Nigeria"],
        "data": [
          [79, 81, 84, 75, 68, 53],
          [80, 82, 83, 76, 71, 55]
        ]
      }
    },
    config = {
      "graphType": "Dumbbell",
      "dumbbellType": "arrow",
      "dumbbellArrowDirection": "order",
      "title": "Life expectancy, 2014 to 2024",
      "legendPosition": "bottom",
      "source": "World Bank"
    },
    width = 600,
    height = 600
  )
)
```
