# Scatter Plot with Regression Line: Data Trend Visualization

Explore a Scatter plot with a regression line for clear data trend visualization.

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/scatter2d-8.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "scatter2d8",
    data = {
      "y": {
        "data": [
          [18, 76.1],
          [19, 77],
          [20, 78.1],
          [21, 78.2],
          [22, 78.8],
          [23, 79.7],
          [24, 79.9],
          [25, 81.1],
          [26, 81.2],
          [27, 81.8],
          [28, 82.8],
          [29, 83.5]
        ],
        "smps": ["Age", "Height"],
        "vars": ["s1", "s2", "s3", "s4", "s5", "s6", "s7", "s8", "s9", "s10", "s11", "s12"]
      }
    },
    config = {
      "citation": "Moore, David S., and George P. McCabe (1989)",
      "citationScaleFontFactor": 0.75,
      "graphType": "Scatter2D",
      "showRegressionFit": True,
      "title": "Mean heights of a group of children in Kalama",
      "xAxis": ["Age"],
      "yAxis": ["Height"]
    },
    width = 600,
    height = 600
  )
)
```
