# Multiple donuts: three shares for six entities

Six proteomes as small-multiple donut charts, each split into helix, sheet and coil shares, with the dominant share printed in the hole. Pie graph with pieInnerRadius segregated into a grid.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/donnut-3.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "donnut3",
    data = {
      "y": {
        "vars": ["Alpha helix", "Beta sheet", "Coil"],
        "smps": ["E. coli", "Yeast", "Fruit fly", "Mouse", "Arabidopsis", "Rice"],
        "data": [
          [38, 40, 41, 42, 36, 35],
          [22, 21, 22, 21, 26, 27],
          [40, 39, 37, 37, 38, 38]
        ]
      }
    },
    config = {
      "graphType": "Pie",
      "theme": "minimal",
      "pieInnerRadius": 0.5,
      "pieSegmentLabels": "outside",
      "pieSegmentPrecision": 0,
      "pieSegmentSeparation": 0,
      "showPieGrid": False,
      "showPieSampleLabel": True,
      "colorScheme": "User",
      "colors": ["#0072B2", "#E69F00", "#999999"],
      "xAxis": ["E. coli", "Yeast", "Fruit fly", "Mouse", "Arabidopsis", "Rice"],
      "title": "Protein secondary structure across six proteomes",
      "subtitle": "Share of residues by structure class (synthetic)",
      "legendPosition": "top",
      "showTransition": False
    },
    width = 760,
    height = 560
  )
)
```
