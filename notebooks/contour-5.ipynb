# Smooth Data Contour Plot

A contour plot of a smooth data set shows regions of constant value with a few rows and columns.

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/contour-5.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "contour5",
    data = {
      "y": {
        "data": [
          [16, 17, 18, 19, 19, 21, 22],
          [15, 16, 17, 18, 20, 21, 23],
          [14, 13, 14, 15, 18, 21, 24],
          [13, 12, 11, 13, 16, 23, 25],
          [12, 11, 7, 8, 9, 27, 26],
          [7, 6, 7, 8, 10, 5, 17],
          [4, 1, 2, 3, 4, 10, 14],
          [2, 3, 4, 5, 7, 9, 11]
        ],
        "smps": ["s1", "s2", "s3", "s4", "s5", "s6", "s7"],
        "vars": ["v1", "v2", "v3", "v4", "v5", "v6", "v7", "v8"]
      }
    },
    config = {
      "contourType": "normal",
      "graphType": "Heatmap",
      "heatmapCellBox": False,
      "lineType": "spline",
      "showContourLevel": False,
      "showSampleNames": False,
      "showVariableNames": False,
      "xAxis": ["v1", "v2", "v3", "v4", "v5", "v6", "v7", "v8"]
    },
    after_render = [
      [
        "createContour",
        []
      ]
    ],
    width = 600,
    height = 600
  )
)
```
