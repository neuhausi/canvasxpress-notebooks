# Scatter2D with a categorical axis

A direct-API Scatter2D config can name a categorical column (here, Species) as an axis: the engine promotes that column into the axis and plots one point per observation, instead of the classic sample-vs-sample convention.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/scatter2d-43.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "scatter2d43",
    data = {
      "y": {
        "vars": ["Species", "Sepal.Length", "Sepal.Width"],
        "smps": ["s1", "s2", "s3", "s4", "s5", "s6", "s7", "s8", "s9"],
        "data": [
          [0, 0, 0, 1, 1, 1, 2, 2, 2],
          [5.1, 4.9, 4.7, 6.4, 6.9, 5.5, 7.1, 7.3, 6.7],
          [3.5, 3.0, 3.2, 3.2, 3.1, 2.3, 3.0, 2.9, 2.5]
        ]
      }
    },
    config = {
      "graphType": "Scatter2D",
      "xAxis": ["Species"],
      "yAxis": ["Sepal.Length"],
      "title": "Sepal Length by Species",
      "showLegend": True
    },
    width = 600,
    height = 600
  )
)
```
