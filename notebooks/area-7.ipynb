# Dynamic Area Graph for Data Comparison

Visualize changes in three stacked filtered data series with an area graph in wide format, joined by a spline.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/area-7.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "area7",
    data = {
      "y": {
        "data": [
          [10, 11, 13, 4, 18, 21],
          [5, 6, 10, 12, 15, 10],
          [2, 3, 6, 4, 8, 6]
        ],
        "smps": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
        "vars": ["Series A", "Series B", "Series C"]
      }
    },
    config = {
      "areaType": "stacked",
      "colorScheme": "LastAirBenderWater",
      "filterData": [
        [
          "guess",
          False,
          "exact",
          ["Series B"]
        ]
      ],
      "graphOrientation": "vertical",
      "graphType": "Area",
      "lineType": "spline",
      "smpTextRotate": 90,
      "smpTitle": "Month / First Quarters of 2024",
      "subtitle": "Filtered data",
      "title": "Area graph with three stacked data series",
      "titleScaleFontFactor": 1.2,
      "xAxis": ["Series A", "Series B", "Series C"],
      "xAxisTitle": "Revenue (in Millions)"
    },
    width = 600,
    height = 600
  )
)
```
