# Sparklines — win/loss bars with summary value

A win/loss sparkline reduces each series to equal-height up/down bars coloured by sign, so the run of gains and losses reads at a glance; the right-edge label shows each region's mean quarterly change.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/sparklines-2.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "sparklines2",
    data = {
      "y": {
        "vars": ["QoQ Change"],
        "smps": ["Q1-N", "Q2-N", "Q3-N", "Q4-N", "Q5-N", "Q6-N", "Q1-S", "Q2-S", "Q3-S", "Q4-S", "Q5-S", "Q6-S", "Q1-E", "Q2-E", "Q3-E", "Q4-E", "Q5-E", "Q6-E", "Q1-W", "Q2-W", "Q3-W", "Q4-W", "Q5-W", "Q6-W"],
        "data": [
          [3.2, -1.5, 2.1, 4.0, -0.8, 1.9, -2.1, 1.2, -3.0, 0.5, 2.2, -1.1, 5.0, 4.1, -2.2, 3.3, 1.0, 2.8, -1.0, -2.5, 1.1, -0.4, 0.9, -1.8]
        ]
      },
      "x": {
        "Region": ["North", "North", "North", "North", "North", "North", "South", "South", "South", "South", "South", "South", "East", "East", "East", "East", "East", "East", "West", "West", "West", "West", "West", "West"]
      }
    },
    config = {
      "graphType": "Sparklines",
      "sparklineType": "winloss",
      "sparklineAggregate": "mean",
      "title": "Quarterly revenue change by region (win / loss)"
    },
    width = 600,
    height = 600
  )
)
```
