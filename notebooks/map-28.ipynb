# Dorling cartogram — US population

A Dorling cartogram: each state becomes a circle sized by its population and nudged so none overlap, while staying near its true location. Set <b>mapType</b> to <b>dorling</b> with a <b>sizeBy</b> variable.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/map-28.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).
