# Interactive Area-Line Graph

Area-line combination graph visualizes multiple series in each axis, emphasizing trends with a spline connection.

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/arealine-1.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "arealine1",
    data = {
      "y": {
        "data": [
          [10, 11, 13, 4, 18, 21],
          [5, 8, 10, 12, 15, 10],
          [2, 3, 6, 4, 8, 6],
          [18, 27, 29, 21, 42, 37],
          [6, 9, 9, 6, 14, 12]
        ],
        "smps": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
        "vars": ["Series A", "Series B", "Series C", "Series D", "Series E"]
      }
    },
    config = {
      "graphOrientation": "vertical",
      "graphType": "AreaLine",
      "lineThickness": 3,
      "lineType": "spline",
      "objectColorTransparency": 0.5,
      "setMaxX": 50,
      "setMaxX2": 50,
      "smpTextRotate": 90,
      "smpTitle": "Month",
      "xAxis": ["Series A", "Series B", "Series C"],
      "xAxis2": ["Series D", "Series E"],
      "xAxisShow": True,
      "xAxis2Show": True
    },
    width = 600,
    height = 600
  )
)
```
