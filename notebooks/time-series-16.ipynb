# Seven yearly values with a value label on each column

Seven annual totals as columns with the value printed above each, a sharp rise then a partial fall, year labels along the bottom and no y axis. Bar graph with data labels on an annual series.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/time-series-16.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "timeseries16",
    data = {
      "y": {
        "vars": ["Hectares"],
        "smps": ["2016 ", "2017 ", "2018 ", "2019 ", "2020 ", "2021 ", "2022 "],
        "data": [
          [18294, 23772, 46262, 86050, 161650, 163542, 66757]
        ]
      }
    },
    config = {
      "graphType": "Bar",
      "titleAlign": "plot::0",
      "subtitleAlign": "plot::0",
      "legendHorizontalJustification": 0,
      "legendJustifyTo": "plot",
      "graphOrientation": "vertical",
      "theme": "classic",
      "xAxisGridMajorShow": True,
      "colorScheme": "User",
      "colors": ["#C8442E"],
      "xAxis": ["Hectares"],
      "showLegend": False,
      "title": "In 7 years, 620,000 hectares of wetland drained",
      "subtitle": "Annual wetland area drained in the fictional Aldermere basin, in hectares.",
      "xAxisNumberFormat": "compact",
      "smpTextRotate": 90,
      "smpTitle": "",
      "xAxisTitle": "",
      "sampleSpaceFactor": 0.7
    },
    width = 760,
    height = 480
  )
)
```
