# Excluded Controls (Outliers Flagged)

Plate-read control values with three excluded outliers flagged by a red X, placed with a sample-anchored text decoration.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/line-8.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "line8",
    data = {
      "y": {
        "vars": ["Replication01", "Replication02"],
        "smps": ["A01", "B01", "C01", "D01", "E01", "F01", "G01", "H01", "I01", "J01", "K01", "L01", "M01", "N01", "O01", "P01"],
        "data": [
          [0.22, 0.227, 0.212, 0.655, 0.211, 0.204, 0.2, 0.202, 0.189, 0.196, 0.199, 0.201, 0.184, 0.196, 0.185, 0.207],
          [0.213, 0.22, 0.208, 0.221, 0.211, 0.207, 0.198, 0.198, 0.04, 0.194, 0.189, 0.201, 0.189, 0.192, 1.018, 0.199]
        ]
      }
    },
    config = {
      "graphType": "Line",
      "graphOrientation": "vertical",
      "lineType": "spline",
      "tension": 0.2,
      "dataPointSize": 20,
      "lineDecoration": "dot",
      "lineThickness": 2,
      "colors": ["#C1C1C1", "#FF8C00"],
      "smpTitle": "Wells",
      "smpTextRotate": 45,
      "xAxisTitle": "Control value",
      "title": "Excluded Controls (Outliers Flagged)",
      "showLegend": True,
      "showSampleNames": True,
      "showDecorations": True,
      "decorations": {
        "text": [
          {
            "smp": "D01",
            "var": "Replication01",
            "label": "✕",
            "color": "rgb(220,0,0)"
          },
          {
            "smp": "I01",
            "var": "Replication02",
            "label": "✕",
            "color": "rgb(220,0,0)"
          },
          {
            "smp": "O01",
            "var": "Replication02",
            "label": "✕",
            "color": "rgb(220,0,0)"
          }
        ]
      }
    },
    width = 600,
    height = 450
  )
)
```
