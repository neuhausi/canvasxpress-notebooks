# 3 Series Stacked Wide Format Spline Area Graph

An informative wide-format area graph showcasing three stacked data series connected by a spline for clear trend visualization.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/area-4.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "area4",
    data = {
      "y": {
        "data": [
          [10, 11, 13, 4, 18, 21],
          [5, 6, 10, 12, 15, 10],
          [2, 3, 6, 4, 8, 6]
        ],
        "smps": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
        "vars": ["Series A", "Series B", "Series C"]
      }
    },
    config = {
      "areaType": "stacked",
      "colorScheme": "LastAirBenderWater",
      "graphOrientation": "vertical",
      "graphType": "Area",
      "lineType": "spline",
      "smpTextRotate": 90,
      "smpTitle": "Month / First Quarters of 2024",
      "title": "Area graph with three stacked data series",
      "titleScaleFontFactor": 1.2,
      "xAxis": ["Series A", "Series B", "Series C"],
      "xAxisTitle": "Revenue (in Millions)"
    },
    width = 600,
    height = 600
  )
)
```
