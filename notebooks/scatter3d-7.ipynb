# 3D Scatter Plot for Complex Relationships

Visualize complex relationships between three variables with a 3D scatter plot.

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/scatter3d-7.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "scatter3d7",
    data = {
      "fx": "sin(2y) * cos(3x) / sec(xy)"
    },
    config = {
      "colorBy": "Y",
      "graphType": "Scatter3D",
      "outline3DColor": "rgba(0,0,0,0.1)"
    },
    width = 600,
    height = 600
  )
)
```
