# Tiled discrete heatmap

A discrete-colour, padded-tile heatmap via heatmapType:'tiled' (gaps between cells + binned colours + no cell box). Five value classes (colorSpectrumBreaks) are named a-e in a stepped legend (colorLegendLabels), and the square matrix of equally sized tiles reads like an oncoprint - ideal for a compact perturbation/landscape view across samples and variables.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/heatmap-22.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).
