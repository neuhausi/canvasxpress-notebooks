# Gauge with Target

A gauge meter with a target/reference marker: the needle shows the value (Revenue) while the tick marks the target (Goal), summarized on the same basis — the same target concept bullet graphs use.

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/meter-24.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "meter24",
    data = {
      "y": {
        "vars": ["Revenue", "Goal"],
        "smps": ["Q3"],
        "data": [
          [82],
          [95]
        ]
      }
    },
    config = {
      "graphType": "Meter",
      "meterType": "gauge",
      "summaryType": "sum",
      "xAxis": ["Revenue"],
      "bulletTargetVarName": "Goal",
      "bulletTargetVar": "rgb(20,20,20)",
      "rangeSegments": [0, 25, 50, 75, 100],
      "setMin": 0,
      "setMax": 100
    },
    width = 600,
    height = 600
  )
)
```
