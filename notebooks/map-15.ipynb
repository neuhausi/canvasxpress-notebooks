# Customizable Leaflet Map with Interactive Markers

Create an interactive Leaflet map with customizable markers to visualize geographical data efficiently.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/map-15.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "map15",
    data = {
      "x": {
        "Latitude": [40.701, 40.702, 40.703],
        "Longitude": [-73.985, -74.011, -73.991],
        "color": ["blue", "red", "green"],
        "shape": ["circle", "teardrop", "star"],
        "size": [1, 1.5, 3]
      },
      "y": {
        "smps": ["New York 1", "New York 2", "New York 3"]
      }
    },
    config = {
      "graphType": "Map",
      "mapConfig": {
        "center": [40.7, -74],
        "zoom": 15
      },
      "markerBy": "shape",
      "showLegend": False,
      "useLeaflet": True
    },
    width = 600,
    height = 600
  )
)
```
