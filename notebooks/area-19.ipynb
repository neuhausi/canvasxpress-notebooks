# TrendPlot (stacked bars over a stacked area)

A TrendPlot draws a solid stacked bar at each category on top of a semi-transparent stacked area. The bars preserve category separation while the area makes the across-category trend easy to read. Set areaShowColumns:true on any Area graph; areaColumnWidth controls the bar width, and the overlay composes with any areaType (stacked, percent, overlapping) and areaStyle.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/area-19.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "area19",
    data = {
      "y": {
        "vars": ["Mobile", "Desktop", "Tablet"],
        "smps": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
        "data": [
          [32, 38, 41, 45, 52, 58],
          [40, 42, 39, 44, 41, 43],
          [12, 14, 16, 15, 18, 20]
        ]
      }
    },
    config = {
      "graphType": "Area",
      "graphOrientation": "vertical",
      "areaType": "stacked",
      "areaStyle": "translucent",
      "areaShowColumns": True,
      "areaColumnWidth": 0.6,
      "colorScheme": "CanvasXpress",
      "title": "TrendPlot: sessions by platform",
      "subtitle": "Stacked bars with a semi-transparent stacked area behind them",
      "smpTitle": "Month",
      "xAxisTitle": "Sessions (thousands)",
      "xAxis": ["Mobile", "Desktop", "Tablet"]
    },
    width = 600,
    height = 600
  )
)
```
