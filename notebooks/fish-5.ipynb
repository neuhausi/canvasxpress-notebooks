# Simplified Fish Plots for Tumor Evolution Visualization

Visualize tumor evolution with simplified fish plots for clear understanding.

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/fish-5.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "fish5",
    data = {
      "y": {
        "data": [
          [100, 2, 2, 98],
          [45, 0, 0, 0],
          [0, 0, 2, 95],
          [0, 0, 1, 40]
        ],
        "smps": [0, 30, 75, 150],
        "vars": ["Founding", "Subclone 1", "Subclone 2", "Subclone 3"]
      }
    },
    config = {
      "backgroundType": "panel",
      "colors": ["rgb(136,136,136)", "rgb(239,0,0)", "rgb(255,96,0)", "rgb(255,207,0)", "rgb(191,255,64)", "rgb(80,255,175)", "rgb(0,223,255)", "rgb(0,112,255)", "rgb(0,0,255)", "rgb(0,0,143)"],
      "fishAxis": [0, 150],
      "fishCloneLabels": ["TP53,MET", "NF1", "", "8q+,6p-"],
      "fishParents": [0, 1, 1, 3],
      "fishShape": "spline",
      "fishTimepoints": [0, 30, 75, 150],
      "graphType": "Fish",
      "legendColumns": 4,
      "legendKeyBackgroundBorderColor": "rgba(255,255,255,0)",
      "legendKeyBackgroundColor": "rgba(255,255,255,0)",
      "legendPosition": "bottom",
      "panelBackgroundColor": "#ffe4c4",
      "xAxis": ["Founding", "Subclone 1", "Subclone 2", "Subclone 3"]
    },
    width = 600,
    height = 450
  )
)
```
