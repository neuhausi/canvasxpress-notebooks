# Visualizing Z Values with Contour Plot

An example of colored contour plot showing z values for each row and column combination.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/contour-2.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "contour2",
    data = {
      "y": {
        "data": [
          [20, 15.625, 12.5, 10.625, 10],
          [15.625, 11.25, 8.125, 6.25, 5.625],
          [12.5, 8.125, 5, 3.125, 2.5],
          [10.625, 6.25, 3.125, 1.25, 0.625],
          [10, 5.625, 2.5, 0.625, 0]
        ],
        "smps": ["s1", "s2", "s3", "s4", "s5"],
        "vars": ["v1", "v2", "v3", "v4", "v5"]
      }
    },
    config = {
      "contourType": "normal",
      "graphType": "Heatmap",
      "heatmapCellBox": False,
      "showSampleNames": False,
      "showVariableNames": False,
      "title": "Basic Contour Plot",
      "xAxis": ["v1", "v2", "v3", "v4", "v5"]
    },
    after_render = [
      [
        "createContour",
        []
      ]
    ],
    width = 600,
    height = 600
  )
)
```
