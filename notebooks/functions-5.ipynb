# 3D Function Scatter Plot: A Visual Representation

A 3D function scatter plot visually displays a colored graph representing mathematical functions in three dimensions.

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/functions-5.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "functions5",
    data = {
      "fx": "sin(2y) * cos(3x) / sec(xy)"
    },
    config = {
      "colorBy": "Y",
      "graphType": "Scatter3D",
      "outline3DColor": "rgba(0,0,0,0.1)"
    },
    width = 600,
    height = 600
  )
)
```
