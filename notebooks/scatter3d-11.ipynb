# High-resolution 3D function surface

A smooth analytic surface z = sin(2x)·cos(2y) plotted from a function (fx). fxSamples sets the grid resolution (80 subdivisions here) for a fine, well-coloured mesh, and aspect3D:'data' scales the box to the real per-axis data spans so the surface keeps true data proportions.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/scatter3d-11.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "scatter3d11",
    data = {
      "fx": "sin(2 * x) * cos(2 * y)"
    },
    config = {
      "graphType": "Scatter3D",
      "fxSamples": 80,
      "aspect3D": "data",
      "colorBy": "Y",
      "colorSpectrum": ["#2c7bb6", "#abd9e9", "#ffffbf", "#fdae61", "#d7191c"],
      "xRotate": 45,
      "yRotate": 35,
      "outline3DColor": "rgba(0,0,0,0)",
      "smoothSurface": True,
      "title": "High-resolution 3D Function Surface"
    },
    width = 600,
    height = 600
  )
)
```
