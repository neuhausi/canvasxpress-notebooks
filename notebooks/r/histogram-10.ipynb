# Master Histograms for Data Analysis

Understanding histograms: a powerful visual tool to analyze data distributions by grouping data into bins and displaying frequencies, highlighting patterns effectively.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/histogram-10.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-mtcars2-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
z=read.table("https://www.canvasxpress.org/data/r/cX-mtcars2-var.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
varAnnot=z,
graphType="Scatter2D",
histogramAsDotplot=TRUE,
histogramBinWidth=3,
showHistogram=TRUE,
xAxis=list("mpg"),
xAxisTitle="mpg",
yAxisTitle="Frequency"
)
```
