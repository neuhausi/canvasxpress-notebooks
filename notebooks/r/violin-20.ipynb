# Half Violin with Boxplot

The half violin and its boxplot occupy separate partitions of the band instead of overlapping.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/violin-20.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-violin20-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
x=read.table("https://www.canvasxpress.org/data/r/cX-violin20-smp.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
smpAnnot=x,
axisAlgorithm="rPretty",
axisTitleFontStyle="bold",
background="white",
backgroundType="panel",
graphOrientation="vertical",
graphType="Boxplot",
groupingFactors=list("dose"),
guidesColor="white",
guidesLineType="solid",
guidesShow=TRUE,
panelBackgroundColor="#E5E5E5",
showBoxplotIfViolin=TRUE,
showLegend=FALSE,
showViolinBoxplot=TRUE,
smpTextRotate=90,
smpTitle="dose",
smpTitleFontStyle="bold",
title="Half Violin with Boxplot",
violinScale="area",
violinSide="lt",
xAxis=list("len"),
xAxis2Show=FALSE,
xAxisGridMajorColor="white",
xAxisGridMinorShow=FALSE,
xAxisTitle="len"
)
```
