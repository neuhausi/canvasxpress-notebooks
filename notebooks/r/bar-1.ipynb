# Simple Bar Graph for Easy Comparison

A basic bar graph displaying a single series for easy comparison.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/bar-1.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-stBarOneSeries-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
graphOrientation="vertical",
graphType="Bar",
showLegend=FALSE,
smpTextRotate=90,
smpTitle="Categories",
title="Bar graph with a single series",
xAxis=list("Var 1"),
xAxisTitle="Var 1"
)
```
