# WikiPathways Network Graph Visualization

Visualizing the interconnected entities and relationships from Wikipathways using a network graph.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/network-17.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
nodes=read.table("https://www.canvasxpress.org/data/r/cX-wpapoptosis-nodes.txt", header=TRUE, sep="\t", quote="", fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
edges=read.table("https://www.canvasxpress.org/data/r/cX-wpapoptosis-edges.txt", header=TRUE, sep="\t", quote="", fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
nodeData=nodes,
edgeData=edges,
adjustBezier=FALSE,
calculateLayout=FALSE,
graphType="Network",
networkFreeze=TRUE,
networkNodesOnTop=FALSE,
preScaleNetwork=FALSE,
showNodeNameThreshold=20000,
title="Apoptosis"
)
```
