# Advantages of Genome Graphs in Diagnostics

Genome graphs offer a comprehensive view of genetic variation within populations, enhancing disease diagnostics and personalized medicine.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/genome-1.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
library(jsonlite)
genome <- read_json("https://www.canvasxpress.org/data/r/cX-genomesimple.json")
canvasXpress(
data=genome,
background="rgb(245,245,245)",
graphType="Genome",
setMax=30,
setMin=0
)
```
