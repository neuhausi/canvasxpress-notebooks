# Master Data Visualization with Dot Plots

Dot plots are ideal for visualizing data distributions, offering a clear and intuitive way to show the frequency of individual data points.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/dotplot-4.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-dumbbell-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
axisAlgorithm="wilkinson",
dataPointSizeScaleFactor=3,
dataTextScaleFontFactor=0.6,
dotplotType="stacked",
graphType="Dotplot",
plotStyle="open",
showDataValues=TRUE,
smpTitle="School",
sortDir="descending",
title="Gender Earnings Disparity",
xAxis=list("Women", "Men"),
xAxis2Title="Annual Salary",
xAxisGridMinorShow=FALSE,
xAxisTickFormat="$%sK",
xAxisTitle="Annual Salary",
afterRender=list(list("sortSamplesByVariable", list("Men")))
)
```
