# Data Visualization: Grouped Bar Graph

Illustration of a black and white bar graph presenting data in grouped samples.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/bar-5.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-iris-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
x=read.table("https://www.canvasxpress.org/data/r/cX-iris-smp.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
smpAnnot=x,
graphOrientation="vertical",
graphType="Bar",
groupingFactors=list("Species"),
layoutTopology="1X3",
legendColumns=2,
legendPosition="bottom",
segregateSamplesBy=list("Species"),
smpTextRotate=90,
smpTitle="Species",
theme="blackAndWhite",
title="Iris flower data set",
xAxis=list("Sepal.Length", "Sepal.Width", "Petal.Length", "Petal.Width")
)
```
