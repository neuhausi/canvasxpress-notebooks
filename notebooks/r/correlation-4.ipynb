# Visualizing correlations in mtcars data

Explore the correlation of variables in the mtcars dataset using correlation graphs.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/correlation-4.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-mtcarst-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
graphType="Heatmap",
title="Heatmap - Correlation",
xAxis=list("mpg", "cyl", "disp", "hp", "drat", "wt", "qsec", "vs", "am", "gear", "carb"),
afterRender=list(list("createHeatmapCorrelation", list()), list("clusterVariables", list()), list("clusterSamples", list()))
)
```
