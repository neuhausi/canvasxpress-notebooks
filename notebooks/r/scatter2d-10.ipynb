# Non-linear Scatter Plot with Line Decoration

Scatter plot featuring non-linear fit decoration with lines included in data object, not configuration.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/scatter2d-10.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-nonlinearfit-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
decorations=list(nlfit=list(list(label="Custom Fit", param=list(164, 313, 0.031, -1.5, 1.2e-06, 1.9), type="cst"), list(label="Regular Fit", param=list(164, 313, 0.031, 1.5, 1.2e-06, 1.9), type="reg"))),
graphType="Scatter2D",
setMaxY=350,
setMinY=100,
showDecorations=TRUE,
xAxis=list("Concentration"),
xAxisTransform="log10",
xAxisTransformTicks=TRUE,
yAxis=list("V1"),
yAxisExact=TRUE
)
```
