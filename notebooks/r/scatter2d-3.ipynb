# Interactive Scatter2D Plot with Loess Fit Line

Interactive Scatter2D graph displaying data points with a loess fit line and confidence intervals.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/scatter2d-3.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-loess2-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
graphType="Scatter2D",
showConfidenceIntervals=TRUE,
showLoessFit=TRUE,
title="Loess Fit",
xAxis=list("E"),
yAxis=list("NOx")
)
```
