# Visualizing Trends in USpopage with Area Graph

Explore a visually captivating Area Graph with multiple Proportional Stacked data series, showcasing trends over time from the R dataset USpopage.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/area-12.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-area-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
areaType="percent",
colorScheme="GGPlot",
graphOrientation="vertical",
graphType="Area",
smpLabelInterval=20,
smpTextRotate=45,
smpTitle="Year",
subtitle="gcookbook - uspopage",
title="Age distribution of population in the United States",
xAxis=list("<5", "5-14", "15-24", "25-34", "35-44", "45-54", "55-64", ">64"),
xAxisTitle="Number of People (1000's)"
)
```
