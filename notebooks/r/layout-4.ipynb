# Divide & Compare: Facets in Data Visualization

Facets in Data Visualization splits data into subsets, creating multiple plots to compare trends side by side for clearer insights.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/layout-4.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-irist-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
z=read.table("https://www.canvasxpress.org/data/r/cX-irist-var.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
varAnnot=z,
broadcast=TRUE,
colorBy="Species",
graphType="Scatter2D",
layoutAdjust=FALSE,
scatterPlotMatrix=TRUE,
xAxis=list("Sepal.Length", "Sepal.Width", "Petal.Length", "Petal.Width")
)
```
