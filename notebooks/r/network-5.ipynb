# Force-Directed Cola Layout Algorithm: Explained

Network graph visualization algorithm Force-Directed Cola Layout optimizes node positioning based on physics principles.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/network-5.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
nodes=read.table("https://www.canvasxpress.org/data/r/cX-miserables-nodes.txt", header=TRUE, sep="\t", quote="", fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
edges=read.table("https://www.canvasxpress.org/data/r/cX-miserables-edges.txt", header=TRUE, sep="\t", quote="", fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
nodeData=nodes,
edgeData=edges,
colorNodeBy="group",
graphType="Network",
is3DNetwork=TRUE,
networkColaJaccardLinkLengthDefault=0.7,
networkColaJaccardLinkLengths=40,
networkColaStartUnconstrainedIterations=50,
networkLayoutType="cola"
)
```
