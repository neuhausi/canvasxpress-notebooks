# Customizable Interactive Map Data Analysis

Explore interactive and customizable maps with embedded data for detailed spatial analysis and visualization.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/map-1.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-OlympicMedalsT-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
colorBy="Total",
graphType="Map",
legendPosition="bottom",
mapId="medals",
theme="tableau",
title="Total Number of Olympic Medals in Paris - 2024",
topoJSON="https://www.canvasxpress.org/data/maps/WORLD.json"
)
```
