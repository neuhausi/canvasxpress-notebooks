# Visual Data Comparison with Lollipop Graph

Eye-catching lollipop graph perfect for comparing data visually.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/lollipop-1.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-lollipop-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
x=read.table("https://www.canvasxpress.org/data/r/cX-lollipop-smp.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
smpAnnot=x,
barType="lollipop",
colorScheme="CanvasXpress",
dataPointSizeScaleFactor=6,
graphType="Bar",
sizeBy="val",
widthFactor=0.2,
xAxis=list("V1")
)
```
