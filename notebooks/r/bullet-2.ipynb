# Visualization in A 1x6 Bullet Graph Layout

Bullet graphs in a 1x6 layout using adjusted scales, target markers, and qualitative ranges to clearly visualize performance.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/bullet-2.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-bullet-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
bulletTargetVarColor="#3F3F3F",
bulletTargetVarName="V5",
colors=list("#3F3F3F"),
graphOrientation="vertical",
graphType="Bullet",
layoutAdjust=TRUE,
layoutTopology="1X6",
rangeColors=list("#945D55", "#C4A285", "#EBE7DE"),
rangeStack=list("V2", "V3", "V4"),
segregateSamplesBy=list("sample"),
showDataValues=TRUE,
smpTextRotate=90,
stripShow=FALSE,
xAxis=list("V1"),
xAxisTextScaleFontFactor=0.6
)
```
