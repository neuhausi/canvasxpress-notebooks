# CodeFlower: Visualize Source Code Connections

Visualize source code connections with CodeFlower to understand project structure.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/network-26.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
nodes=read.table("https://www.canvasxpress.org/data/r/cX-faker-nodes.txt", header=TRUE, sep="\t", quote="", fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
edges=read.table("https://www.canvasxpress.org/data/r/cX-faker-edges.txt", header=TRUE, sep="\t", quote="", fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
nodeData=nodes,
edgeData=edges,
edgeColor="rgb(158,202,225)",
graphType="Network",
useBarnesHutSimulation=TRUE
)
```
