# Source Code Relationship Visualizer with jQuery

Network graph visualization tool designed for showcasing source code relationships using jQuery.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/network-27.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
nodes=read.table("https://www.canvasxpress.org/data/r/cX-jquery-nodes.txt", header=TRUE, sep="\t", quote="", fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
edges=read.table("https://www.canvasxpress.org/data/r/cX-jquery-edges.txt", header=TRUE, sep="\t", quote="", fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
nodeData=nodes,
edgeData=edges,
edgeColor="rgb(158,202,225)",
graphType="Network",
useBarnesHutSimulation=TRUE
)
```
