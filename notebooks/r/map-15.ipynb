# Customizable Leaflet Map with Interactive Markers

Create an interactive Leaflet map with customizable markers to visualize geographical data efficiently.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/map-15.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
x=read.table("https://www.canvasxpress.org/data/r/cX-mapdecor-smp.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
smpAnnot=x,
graphType="Map",
mapConfig=list(center=list(40.7, -74), zoom=15),
markerBy="shape",
showLegend=FALSE,
useLeaflet=TRUE
)
```
