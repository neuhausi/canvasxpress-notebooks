# Ring Meters with Corner Titles

Ring meters with meterRingTitlePosition set to topLeft place each tile's title in the card's top-left corner instead of centered above the circle — here the average satisfaction score per department.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/meter-23.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-meter23-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
x=read.table("https://www.canvasxpress.org/data/r/cX-meter23-smp.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
smpAnnot=x,
graphType="Meter",
groupingFactors=list("Department"),
meterCard=TRUE,
meterRingTitlePosition="topLeft",
meterType="ring",
meterVar="Satisfaction",
rangeColors=list("rgb(38,166,154)"),
summaryType="average",
xAxis=list("Satisfaction")
)
```
