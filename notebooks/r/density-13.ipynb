# Sepal Length Density in Iris Dataset

Density graphs visually represent the distribution of sepal lengths in the Iris dataset, showcasing the concentration of data points and highlighting any notable patterns or outliers.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/density-13.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-irist-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
z=read.table("https://www.canvasxpress.org/data/r/cX-irist-var.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
varAnnot=z,
colorBy="Species",
graphType="Scatter2D",
hideHistogram=TRUE,
histogramStat="count",
segregateVariablesBy=list("Species"),
showFilledHistogramDensity=TRUE,
showHistogram="Species",
showHistogramDensity=TRUE,
showHistogramMedian=FALSE,
xAxis=list("Sepal.Length", "Sepal.Width", "Petal.Length", "Petal.Width")
)
```
