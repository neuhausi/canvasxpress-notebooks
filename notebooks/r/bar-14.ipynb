# Audrey Hepburn Awards: Stacked Bar Graph

An organized stacked bar graph of awards won by Audrey Hepburn showcases multiple series of data effectively.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/bar-14.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-audrey2-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
x=read.table("https://www.canvasxpress.org/data/r/cX-audrey2-smp.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
smpAnnot=x,
colorBy="Result",
colorScheme="Greens",
graphType="Bar",
groupingFactors=list("Award"),
legendColumns=2,
legendPosition="top",
legendTextScaleFontFactor=1.5,
maxSmpStringLen=50,
objectBorderColor="rgb(0,0,0)",
showLegendTitle=FALSE,
smpTextScaleFontFactor=1.5,
stackBy="Result",
summaryType="count",
title="The Awards of Audrey Hepburn",
xAxis=list("Year"),
xAxis2Show=TRUE,
xAxisShow=FALSE
)
```
