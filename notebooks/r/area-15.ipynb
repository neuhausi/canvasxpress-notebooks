# Visualizing Trends: Non-Uniform Area Graph

Area graph with non-uniform data points across two series showcases trends effectively, emphasizing cumulative values for each category.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/area-15.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-scatterArea-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
z=read.table("https://www.canvasxpress.org/data/r/cX-scatterArea-var.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
varAnnot=z,
colorBy="g",
colorScheme="GGPlot",
graphType="Area",
theme="GGPlot",
xAxis=list("x"),
yAxis=list("y")
)
```
