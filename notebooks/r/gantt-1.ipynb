# Efficient Project Planning with Gantt Chart

Visualize project timelines and dependencies effectively with a simple Gantt chart. 

Live example, data and code in other languages: https://www.canvasxpress.org/examples/gantt-1.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-gantt-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
blockContrast=TRUE,
ganttEnd="End",
ganttStart="Start",
graphType="Gantt",
xAxis=list("Start", "End")
)
```
