# Bullet Graph: Performance at a Glance

Bullet graph displaying data value within colored qualitative ranges and including a target marker for quick performance comparison.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/bullet-1.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-bullet-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
bulletTargetVarName="V5",
dataTextColor="#000000",
dataValuesPosition="inside",
graphOrientation="horizontal",
graphType="Bullet",
rangeColors=list("#777777", "#AAAAAA", "#DDDDDD"),
rangeStack=list("V2", "V3", "V4"),
showDataValues=TRUE,
showLegend=FALSE,
xAxis=list("V1")
)
```
