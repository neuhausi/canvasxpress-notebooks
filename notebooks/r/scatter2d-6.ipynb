# Advanced Scatter Plot Analysis with Confidence

Visualize multiple series in a Scatter plot with correlation confidence, showcasing the full range of relationships.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/scatter2d-6.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-mtcars-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
asSampleFactors=list("cyl"),
colorBy="cyl",
confidenceIntervalColorCoordinate=TRUE,
graphType="Scatter2D",
legendBox=TRUE,
showRegressionFit="cyl",
showRegressionFullRange=TRUE,
stringVariableFactors=list("cyl"),
xAxis=list("wt"),
yAxis=list("mpg")
)
```
