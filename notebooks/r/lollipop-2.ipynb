# Colorful Lollipop Chart for Data Visualization

A visually engaging lollipop chart displaying data across categories.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/lollipop-2.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-lollipop2-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
x=read.table("https://www.canvasxpress.org/data/r/cX-lollipop2-smp.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
smpAnnot=x,
barLollipopOpen=TRUE,
barType="lollipopBullet",
colorBy="Color",
colorScheme="GGPlot",
dataPointSizeScaleFactor=7,
graphType="Bar",
marginBottom=50,
marginLeft=50,
marginRight=50,
marginTop=50,
maxTextSize=80,
rangeColors=list("rgb(200,200,200)"),
setMaxX=150,
setMinX=-150,
showDataValues=TRUE,
showLegend=FALSE,
title="Occupations",
xAxis=list("Var1"),
xAxis2Show=FALSE,
xAxisGridMajorShow=FALSE,
xAxisGridMinorShow=FALSE,
xAxisShow=FALSE
)
```
