# Circle Layout: Network Graph Benefits

Circular layout arranges nodes in a network graph in a circle for better visualization and analysis of connections.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/network-13.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
nodes=read.table("https://www.canvasxpress.org/data/r/cX-lesmiserableC-nodes.txt", header=TRUE, sep="\t", quote="", fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
edges=read.table("https://www.canvasxpress.org/data/r/cX-lesmiserableC-edges.txt", header=TRUE, sep="\t", quote="", fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
nodeData=nodes,
edgeData=edges,
background="rgb(245,245,245)",
colorNodeBy="group",
graphType="Network",
networkLayoutType="circular",
nodeScaleFontFactor=1.8,
showAnimation=TRUE,
showLegend=FALSE,
showNodeNameSizeThreshold=25,
showNodeNameThreshold=100,
sizeNodeBy="nodeEdges",
title="Les Miserable"
)
```
