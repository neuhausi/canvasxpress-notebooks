# Ring Meters by Group

When split by a grouping factor, ring meters render one tile per level — here the average miles-per-gallon for 4-, 6- and 8-cylinder cars — each labeled by its cylinder level.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/meter-22.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-meter22-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
x=read.table("https://www.canvasxpress.org/data/r/cX-meter22-smp.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
smpAnnot=x,
graphType="Meter",
groupingFactors=list("cyl"),
meterCard=TRUE,
meterType="ring",
meterVar="mpg",
rangeColors=list("rgb(239,83,80)", "rgb(255,167,38)", "rgb(102,187,106)"),
summaryType="average",
xAxis=list("mpg")
)
```
