# Revolutionizing Genomics with Genome Graphs

Genome graphs revolutionize genomics by integrating diverse genetic variation for enhanced analysis.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/genome-3.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
library(jsonlite)
genome <- read_json("https://www.canvasxpress.org/data/r/cX-genomeadvanced.json")
canvasXpress(
data=genome,
background="rgb(245,245,245)",
dataPointSize=5,
featureStaggered=TRUE,
graphType="Genome"
)
```
