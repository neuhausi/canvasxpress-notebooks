# Dumbbell Chart: Visual Data Comparison

Visualize data relationships using dumbbell graphs that highlight the difference between two points across categories.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/dumbbell-6.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-dumbbell2-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
colors=list("grey", "blue", "pink"),
dumbbellType="arrow",
graphType="Dumbbell",
legendColumns=2,
legendPosition="bottom",
sortDir="ascending",
title="Age Range by Gender",
xAxis=list("Female", "Male"),
xAxis2Show=TRUE,
xAxisShow=FALSE,
xAxisTitle2="Age"
)
```
