# Area Graphs: Effective Data Trend Visualization

Visualize data trends effectively with an Area graph illustrating changes over time, highlighting values through filled areas beneath the line.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/area-1.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-area5-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
colorScheme="Prism",
graphOrientation="vertical",
graphType="Area",
showLegend=FALSE,
smpTextRotate=90,
smpTitle="Month / First Quarters of 2024",
title="Area graph with one series",
titleScaleFontFactor=1.2,
xAxis=list("Value"),
xAxisTitle="Revenue (in Millions)"
)
```
