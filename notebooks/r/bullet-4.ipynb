# Visualization of Bullet Graphs

Clearly visualize performance with bullet graphs, incorporating adjusted scales, target markers, and qualitative ranges.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/bullet-4.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-bullet2-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
bulletTargetVarName="Target",
colors=list("#305DCB", "#E4EFFD", "#ADCFFA", "#73AFF8"),
dataTextColor="#FFFFFF",
dataValuesPosition="inside",
graphOrientation="horizontal",
graphType="Bullet",
rangeColors=list("#73AFF8", "#ADCFFA", "#E4EFFD"),
rangeStack=list("Low", "Average", "High"),
showDataValues=TRUE,
xAxis=list("Value")
)
```
