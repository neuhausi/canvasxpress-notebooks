# Area Line Graphs for Trend Analysis

Area line graphs show trends over time with emphasis on cumulative effect, ideal for financial performance or website traffic analysis.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/arealine-3.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-area8-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
colors=list("rgb(160,81,149)", "rgb(47,75,124)", "rgb(102,81,145)", "rgb(255,166,0)", "rgb(0,63,92)", "rgb(160,81,149)"),
graphOrientation="vertical",
graphType="AreaLine",
legendPosition="topRight",
lineThickness=3,
lineType="spline",
setMinX=0,
setMinX2=0,
smpTextRotate=90,
smpTitle="Month",
xAxis=list("Series A"),
xAxis2=list("Series D"),
xAxis2Show=TRUE,
xAxis2Title="Value2",
xAxisShow=TRUE,
xAxisTitle="Value"
)
```
