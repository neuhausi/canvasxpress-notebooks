# Genome Graphs for Genetic Analysis

Explore the power of genome graphs for comprehensive genetic analysis in sanger sequencing data interpretation.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/genome-2.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
library(jsonlite)
genome <- read_json("https://www.canvasxpress.org/data/r/cX-genomeintermediate.json")
canvasXpress(
data=genome,
background="rgb(245,245,245)",
graphType="Genome",
setMax=30,
setMin=0
)
```
