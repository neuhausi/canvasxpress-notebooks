# Venn Diagrams: Analyzing Data Relationships

Venn diagrams illustrate relationships between sets using overlapping circles, aiding in data comparison and contrast.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/venn-3.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
canvasXpress(
vennData=data.frame(A=340, AB=639, ABC=552, ABCD=148, ABD=578, AC=456, ACD=298, AD=257, B=562, BC=915, BCD=613, BD=354, C=620, CD=143, D=592),
vennLegend=list(A="List 1", B="List 2", C="List 3", D="List 4"),
graphType="Venn",
vennGroups=2
)
```
