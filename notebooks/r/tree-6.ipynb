# Hierarchical Relationships Visualized: Bracket Tree Layout

A bracket plot visualizing hierarchical relationships through a tree layout.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/tree-6.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-bracket-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
x=read.table("https://www.canvasxpress.org/data/r/cX-bracket-smp.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
smpAnnot=x,
graphType="Tree",
hierarchy=list("Final", "Semifinal", "4th", "8th"),
showTransition=FALSE,
title="Bracket",
treeBracketLabelAlign="left",
treeClickDisable=TRUE,
treeInverted=TRUE,
treeNodeSizeScaleFactor=4,
treeType="bracket",
xAxis=list("Goals")
)
```
