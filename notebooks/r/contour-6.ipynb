# Master Density Contour Graphs: A Visual Data Analysis Tool

Learn how to interpret density contour graphs, a powerful visualization tool for understanding data distribution patterns.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/contour-6.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-contourdensity-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
graphType="Contour",
showContourDataPoints=TRUE,
showContourLevel=FALSE,
xAxis=list("eruptions"),
yAxis=list("waiting")
)
```
