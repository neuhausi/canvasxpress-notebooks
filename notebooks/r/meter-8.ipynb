# Interactive Meter Graph for Metadata & Levels

Track metadata and levels in digital format with an interactive meter graph.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/meter-8.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-meter3-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
x=read.table("https://www.canvasxpress.org/data/r/cX-meter3-smp.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
smpAnnot=x,
graphType="Meter",
groupingFactors=list("Quarter"),
meterType="digital",
meterVar="Revenue",
summaryType="sum",
xAxis=list("Revenue")
)
```
