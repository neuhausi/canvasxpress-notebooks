# Visualize Data with Spider Plots

Spider plots, also known as radar charts, display multivariate data in a single graphic, making them ideal for visualizing and comparing multiple variables at once.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/scatter2d-2.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-spider-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
z=read.table("https://www.canvasxpress.org/data/r/cX-spider-var.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
varAnnot=z,
colorBy="Response",
graphType="Scatter2D",
legendBox=TRUE,
legendInside=TRUE,
legendPosition="topRight",
lineBy="Subject",
title="Tumor Response by Week",
xAxis=list("Weeks"),
yAxis=list("Change From Baseline %")
)
```
