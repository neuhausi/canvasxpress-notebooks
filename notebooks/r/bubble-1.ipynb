# Interactive Bubble Plot Analysis

Interactive hierarchical bubble plot showing relationships between variables without grouping them.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/bubble-1.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-CO2-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
x=read.table("https://www.canvasxpress.org/data/r/cX-CO2-smp.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
smpAnnot=x,
bubbleLabel="auto",
circularType="bubble",
colorBy="Continent",
graphType="Circular",
hierarchy=list("Country"),
legendColumns=4,
legendPosition="bottom",
theme="paulTol",
title="Annual CO2 Emmisions in 2018",
xAxis=list("CO2")
)
```
