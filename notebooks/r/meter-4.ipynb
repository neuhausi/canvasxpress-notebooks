# Efficient Data Trend Analysis with Light Meter Graph

Monitor and analyze data trends efficiently with a light meter graph format.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/meter-4.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-meter-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
graphType="Meter",
meterType="state",
xAxis=list("Performance")
)
```
