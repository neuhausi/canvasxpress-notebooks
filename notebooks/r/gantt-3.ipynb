# Grouped Gantt Chart: Efficient Project Timeline Visualization

Visualize project timelines with a grouped Gantt chart for efficient task management and team collaboration.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/gantt-3.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-gantt2-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
x=read.table("https://www.canvasxpress.org/data/r/cX-gantt2-smp.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
smpAnnot=x,
blockContrast=TRUE,
colorBy="Clinical Trial",
ganttCompletion="Completion",
ganttEnd="End",
ganttStart="Start",
graphType="Gantt",
groupingFactors=list("Clinical Trial"),
xAxis=list("Start", "End")
)
```
