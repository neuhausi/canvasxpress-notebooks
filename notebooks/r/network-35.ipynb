# Visualizing Code: Exploring Zend Framework 2 with CodeFlower

Network graphs can visualize relationships in source code, such as the code structure of Zend Framework 2 (zf2) with CodeFlower.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/network-35.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
nodes=read.table("https://www.canvasxpress.org/data/r/cX-zf2-nodes.txt", header=TRUE, sep="\t", quote="", fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
edges=read.table("https://www.canvasxpress.org/data/r/cX-zf2-edges.txt", header=TRUE, sep="\t", quote="", fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
nodeData=nodes,
edgeData=edges,
edgeColor="rgb(158,202,225)",
graphType="Network",
useBarnesHutSimulation=TRUE
)
```
