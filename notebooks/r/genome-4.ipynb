# Genomics Visualization with Ideograms

Genome graphs provide a revolutionary advancement in genomics by integrating diverse genetic variations for more accurate analysis.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/genome-4.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
library(jsonlite)
genome <- read_json("https://www.canvasxpress.org/data/r/cX-genomeideogram.json")
canvasXpress(
data=genome,
background="rgb(245,245,245)",
graphType="Genome",
ideogramShowFullChromosome=TRUE,
showIdeogram=TRUE
)
```
