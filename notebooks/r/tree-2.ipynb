# Efficient Tree Graphs for Data Hierarchy

A tree graph displays hierarchial data, with nodes branching from a root node, enabling efficient search and sort algorithms.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/tree-2.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-tree2-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
x=read.table("https://www.canvasxpress.org/data/r/cX-tree2-smp.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
smpAnnot=x,
graphType="Tree",
hierarchy=list("Level1", "Level2"),
showTransition=TRUE,
title="Collapsible Tree",
xAxis=list("Order")
)
```
