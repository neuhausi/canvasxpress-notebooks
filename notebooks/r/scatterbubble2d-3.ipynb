# Dynamic Data Exploration with ScatterBubble2D Graph

Create an engaging ScatterBubble2D graph with interactive motion filters for dynamic data exploration.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/scatterbubble2d-3.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-bubble-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
z=read.table("https://www.canvasxpress.org/data/r/cX-bubble-var.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
varAnnot=z,
colorBy="Continent",
graphType="ScatterBubble2D",
workflowBy="Year",
xAxis=list("LifeExpectancy"),
yAxis=list("GDPPerCapita"),
yAxisTransform="log2",
zAxis=list("Population")
)
```
