Within each panel, bars are grouped by stress treatment (ABA, Cold, PEG, Salt) and colored by time point (Mock, 6h, 12h, 24h), with sample-anchored error bars drawn as decorations and on-bar data values displayed. Faceting, grouping, custom color mapping, and per-bar error annotations are all driven by configuration alone.
