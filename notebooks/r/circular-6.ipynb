# Relationship Ideogram Visualization: Metadata Categorization

Circular plot with ideogram displaying relationships, categorized by metadata, for effective visualization.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/circular-6.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).
