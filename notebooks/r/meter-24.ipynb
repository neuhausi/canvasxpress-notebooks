# Gauge with Target

A gauge meter with a target/reference marker: the needle shows the value (Revenue) while the tick marks the target (Goal), summarized on the same basis — the same target concept bullet graphs use.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/meter-24.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-meter24-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
bulletTargetVar="rgb(20,20,20)",
bulletTargetVarName="Goal",
graphType="Meter",
meterType="gauge",
rangeSegments=list(0, 25, 50, 75, 100),
setMax=100,
setMin=0,
summaryType="sum",
xAxis=list("Revenue")
)
```
