# Car Quantity by Category Graph

Bar graph displaying the quantity of cars per category grouped by metadata.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/bar-11.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-mpg2-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
x=read.table("https://www.canvasxpress.org/data/r/cX-mpg2-smp.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
smpAnnot=x,
colorScheme="GGPlot",
graphOrientation="vertical",
graphType="Bar",
groupingFactors=list("class"),
showDataValues=TRUE,
showLegend=FALSE,
smpTextRotate=90,
summaryType="count",
theme="ggplot",
title="Counting the number of cars by class",
xAxis=list("displ")
)
```
