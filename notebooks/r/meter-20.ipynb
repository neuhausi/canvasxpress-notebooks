# Horizontal Meter Dashboard

One horizontal bar card per group, each with its own value axis, in a dashboard grid.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/meter-20.html

Requires the `canvasXpress` R package (`install.packages("canvasXpress")`, preinstalled on Binder).

```r
library(canvasXpress)
y=read.table("https://www.canvasxpress.org/data/r/cX-meter4-dat.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
x=read.table("https://www.canvasxpress.org/data/r/cX-meter4-smp.txt", header=TRUE, sep="\t", quote="", row.names=1, fill=TRUE, check.names=FALSE, stringsAsFactors=FALSE)
canvasXpress(
data=y,
smpAnnot=x,
graphType="Meter",
groupingFactors=list("cyl"),
meterCard=TRUE,
meterThickness=0.7,
meterType="horizontal",
meterVar="mpg",
summaryType="average",
xAxis=list("mpg")
)
```
