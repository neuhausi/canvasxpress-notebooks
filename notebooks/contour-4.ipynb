# Customized Three-Column Dataset Visualization

Visualizing a customized three-column dataset with unique x and y coordinates, all smoothed using loess interpolation.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/contour-4.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "contour4",
    data = {
      "y": {
        "data": [
          [-9, 0, 10],
          [-9, 1, 5.625],
          [-9, 4, 2.5],
          [-9, 5, 0.625],
          [-9, 7, 0],
          [-6, 0, 10.625],
          [-6, 1, 6.25],
          [-6, 4, 3.125],
          [-6, 5, 1.25],
          [-6, 7, 0.625],
          [-5, 0, 12.5],
          [-5, 1, 8.125],
          [-5, 4, 5],
          [-5, 5, 3.125],
          [-5, 7, 2.5],
          [-3, 0, 15.625],
          [-3, 1, 11.25],
          [-3, 4, 8.125],
          [-3, 5, 6.25],
          [-3, 7, 5.625],
          [-1, 0, 20],
          [-1, 1, 15.625],
          [-1, 4, 12.5],
          [-1, 5, 10.625],
          [-1, 7, 10]
        ],
        "smps": ["s1", "s2", "s3"],
        "vars": ["v1", "v2", "v3", "v4", "v5", "v6", "v7", "v8", "v9", "v10", "v11", "v12", "v13", "v14", "v15", "v16", "v17", "v18", "v19", "v20", "v21", "v22", "v23", "v24", "v25"]
      }
    },
    config = {
      "contourFilled": True,
      "contourType": "normal",
      "graphType": "ScatterBubble2D",
      "title": "Custom Contour Plot",
      "xAxis": ["s1"],
      "yAxis": ["s2"],
      "zAxis": ["s3"]
    },
    after_render = [
      [
        "createContour",
        []
      ]
    ],
    width = 600,
    height = 600
  )
)
```
