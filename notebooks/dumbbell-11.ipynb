# Small-multiple ranges with a highlighted mean dot

Seven age groups in two panels, each row a 95% interval bar with the mean as a highlighted dot, panels segregated by a Wave annotation. Dumbbell graph with three series split into small multiples.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/dumbbell-11.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).
