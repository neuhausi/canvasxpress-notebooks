# Bullet bars: observed versus a guideline with bands

Eight cities as horizontal bullet bars of annual PM2.5 against a guideline target tick, with low, moderate and high qualitative bands behind each bar. Bullet graph with rangeSegments.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/bullet-7.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "bullet7",
    data = {
      "y": {
        "vars": ["PM2.5", "Good", "Moderate", "High", "Guideline"],
        "smps": ["City A", "City B", "City C", "City D", "City E", "City F", "City G", "City H"],
        "data": [
          [8, 14, 22, 11, 29, 17, 6, 35],
          [5, 5, 5, 5, 5, 5, 5, 5],
          [10, 10, 10, 10, 10, 10, 10, 10],
          [20, 20, 20, 20, 20, 20, 20, 20],
          [5, 5, 5, 5, 5, 5, 5, 5]
        ]
      }
    },
    config = {
      "graphType": "Bullet",
      "theme": "bw",
      "graphOrientation": "horizontal",
      "xAxis": ["PM2.5"],
      "rangeStack": ["Good", "Moderate", "High"],
      "rangeColors": ["#4477AA", "#99BBDD", "#DDE8F5"],
      "bulletTargetVarName": "Guideline",
      "colorScheme": "User",
      "colors": ["#222222"],
      "showDataValues": True,
      "dataValuesPosition": "inside",
      "dataTextColor": "#FFFFFF",
      "showLegend": False,
      "title": "Which cities exceed the air-quality guideline?",
      "subtitle": "Annual mean PM2.5 (µg/m³) against guideline (tick) and bands (synthetic)",
      "xAxisTitle": "PM2.5 (µg/m³)"
    },
    width = 760,
    height = 520
  )
)
```
