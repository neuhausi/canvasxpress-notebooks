# Publishing Table with In-cell Bars and Sparklines

Live example, data and code in other languages: https://www.canvasxpress.org/examples/bar-19.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "bar19",
    data = {
      "y": {
        "vars": ["2019", "2020", "2021", "2022", "2023", "Trend", "Population"],
        "smps": ["Germany", "France", "Italy", "Spain", "Poland", "Netherlands"],
        "data": [
          [46, 41, 34, 30, 15, 52],
          [42, 39, 31, 27, 16, 50],
          [44, 41, 34, 29, 17, 55],
          [48, 43, 36, 31, 19, 58],
          [51, 45, 38, 33, 21, 61],
          [0, 0, 0, 0, 0, 0],
          [83200000, 68000000, 58900000, 48100000, 36700000, 17900000]
        ]
      }
    },
    config = {
      "graphType": "Bar",
      "view": "table",
      "dataTableFitHeight": True,
      "dataTableFitWidth": True,
      "title": "GDP per capita (thousand USD)",
      "dataTableColWidth": 120,
      "dataTableRowNumbersShow": False,
      "dataTableColHide": ["2019", "2020", "2021", "2022"],
      "dataTableColRenderer": {
        "2023": "bar",
        "Trend": {
          "type": "sparkline",
          "columns": ["2019", "2020", "2021", "2022", "2023"]
        }
      },
      "dataTableColFormat": {
        "Population": "compact"
      }
    },
    width = 600,
    height = 600
  )
)
```
