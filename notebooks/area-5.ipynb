# Interactive Area Graph with 3 Stacked Data Series

Visualize three 100 percent stacked data series in a wide format area graph, linked together with a spline.

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/area-5.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "area5",
    data = {
      "y": {
        "data": [
          [10, 11, 13, 4, 18, 21],
          [5, 6, 10, 12, 15, 10],
          [2, 3, 6, 4, 8, 6]
        ],
        "smps": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
        "vars": ["Series A", "Series B", "Series C"]
      }
    },
    config = {
      "areaType": "percent",
      "colorScheme": "Prism",
      "graphOrientation": "vertical",
      "graphType": "Area",
      "lineType": "spline",
      "smpTextRotate": 90,
      "smpTitle": "Month / First Quarters of 2024",
      "title": "Area graph with three data series in percentage",
      "titleScaleFontFactor": 1.2,
      "xAxis": ["Series A", "Series B", "Series C"],
      "xAxisTitle": "Percent of Revenue (in Millions)"
    },
    width = 600,
    height = 600
  )
)
```
