# Encode a Categorical Edge Attribute as Line Style

Regulatory networks often need more than one edge channel. Here edge color encodes the direction of regulation (activation vs repression), edge width the coefficient magnitude, and edge line style (solid vs dashed) the supporting evidence type - keeping the arrowhead free for another meaning. Line style is driven by the lineEdgeBy parameter, with per-level styles pinned through lineKey and a matching legend.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/network-36.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).
