# Single smooth area with a shaded period and dotted baseline

One smoothed lake-level curve over the last 12,000 years as a filled area, a shaded highlight period, a dotted present-level reference line and a callout. Area graph with lineType spline and range decorations.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/time-series-11.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "timeseries11",
    data = {
      "y": {
        "vars": ["Lake level above present (m)"],
        "smps": ["12 ka", "11.5 ka", "11 ka", "10.5 ka", "10 ka", "9.5 ka", "9 ka", "8.5 ka", "8 ka", "7.5 ka", "7 ka", "6.5 ka", "6 ka", "5.5 ka", "5 ka", "4.5 ka", "4 ka", "3.5 ka", "3 ka", "2.5 ka", "2 ka", "1.5 ka", "1 ka", "0.5 ka", "0 ka"],
        "data": [
          [34, 31, 26, 21.5, 17.2, 14.6, 12.3, 11.2, 11.6, 10.4, 7.6, 5.2, 3.9, 3.4, 3.6, 4.9, 6.3, 7.5, 8.4, 9, 9.4, 9.8, 10.1, 10.4, 10.6]
        ]
      }
    },
    config = {
      "graphType": "Area",
      "graphOrientation": "vertical",
      "theme": "minimal",
      "lineType": "spline",
      "xAxis": ["Lake level above present (m)"],
      "colorScheme": "User",
      "colors": ["#8FB3C2"],
      "objectColorTransparency": 0.6,
      "title": "Rising and falling lake levels in the Chad basin",
      "subtitle": "Lake surface elevation, meters above the present level",
      "smpTitle": False,
      "smpLabelInterval": 4,
      "smpTextRotate": 90,
      "xAxisTitle": False,
      "showLegend": False,
      "setMinX": 0,
      "setMaxX": 36,
      "marginRight": 22,
      "decorations": {
        "range": [
          {
            "smp": "9.5 ka",
            "smp2": "5.5 ka",
            "color": "rgba(240,160,60,0.22)"
          }
        ],
        "line": [
          {
            "value": 10.5,
            "color": "#1B7F9E",
            "width": 3,
            "label": "Present-day level",
            "align": "left",
            "type": "dotted"
          }
        ],
        "marker": [
          {
            "text": "Shorelines mapped\nduring the humid period",
            "variable": "Lake level above present (m)",
            "sample": "7.5 ka",
            "type": "arrow",
            "x": 0.52,
            "y": 0.32
          },
          {
            "text": "African Humid Period",
            "variable": "Lake level above present (m)",
            "sample": "7.5 ka",
            "type": "text",
            "x": 0.4,
            "y": 0.17,
            "color": "#9A6A1A"
          }
        ]
      },
      "byline": "Chart: CanvasXpress",
      "source": "Synthetic data for illustration"
    },
    width = 760,
    height = 560
  )
)
```
