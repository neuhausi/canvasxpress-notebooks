# Drill-down hierarchy (Region → State → City)

Click a bar to drill into the next annotation level; a breadcrumb walks back and ESC steps up.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/bar-23.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "bar23",
    data = {
      "y": {
        "vars": ["Sales"],
        "smps": ["s1", "s2", "s3", "s4", "s5", "s6", "s7", "s8", "s9", "s10", "s11", "s12"],
        "data": [
          [120, 90, 70, 60, 150, 110, 80, 95, 55, 65, 130, 100]
        ]
      },
      "x": {
        "Region": ["West", "West", "West", "West", "East", "East", "East", "East", "South", "South", "South", "South"],
        "State": ["CA", "CA", "WA", "WA", "NY", "NY", "MA", "MA", "GA", "GA", "FL", "FL"],
        "City": ["LA", "SF", "SEA", "TAC", "NYC", "BUF", "BOS", "CAM", "ATL", "SAV", "MIA", "TPA"]
      }
    },
    config = {
      "graphType": "Bar",
      "groupingFactors": ["Region"],
      "drillHierarchy": ["Region", "State", "City"],
      "summaryType": "sum",
      "graphOrientation": "vertical",
      "colorScheme": "CanvasXpress",
      "showLegend": False
    },
    width = 600,
    height = 600
  )
)
```
