# Round-Robin Tournament Visualization with Chord Graph

A chord graph depicting a round-robin tournament, with colored edges indicating the winners.

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/chord-4.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "chord4",
    data = {
      "y": {
        "data": [
          [0, 8, 8, 8],
          [1, 0, 8, 1],
          [1, 1, 0, 8],
          [1, 8, 1, 0]
        ],
        "smps": ["A", "B", "C", "D"],
        "vars": ["A", "B", "C", "D"]
      }
    },
    config = {
      "chordColor": "largest",
      "chordScaleShow": False,
      "circularArc": 360,
      "circularRotate": 180,
      "circularType": "chord",
      "graphType": "Circular",
      "higlightGreyOut": True,
      "legendKeyBackgroundBorderColor": "rgba(255,255,255,0)",
      "legendKeyBackgroundColor": "rgba(255,255,255,0)",
      "objectBorderColor": "rgb(0,0,0)",
      "title": "Four way Round Robin Tournament"
    },
    width = 600,
    height = 600
  )
)
```
