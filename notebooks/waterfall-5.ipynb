# Waterfall Graph for Project Progress

An illustrative Waterfall graph displaying two data series for clear visualization of project or budget progress.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/waterfall-5.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "waterfall5",
    data = {
      "y": {
        "data": [
          [83, -44, 55, -60, 40, 0],
          [50, 20, -30, 60, -10, 0]
        ],
        "smps": ["S1", "S2", "S3", "S4", "S5", "S6"],
        "vars": ["V1", "V2"]
      }
    },
    config = {
      "graphOrientation": "vertical",
      "graphType": "Waterfall",
      "smpTextRotate": 90,
      "smpTitle": "Samples",
      "title": "Traditional Waterfall",
      "xAxis": ["V1", "V2"],
      "xAxisTitle": "Value"
    },
    width = 600,
    height = 600
  )
)
```
