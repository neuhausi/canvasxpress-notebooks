# Vertical Meter Graph for Data Trends

Visualize data trends over time with a vertical meter graph.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/meter-6.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "meter6",
    data = {
      "y": {
        "data": [
          [85]
        ],
        "smps": ["January"],
        "vars": ["Performance"]
      }
    },
    config = {
      "graphType": "Meter",
      "setMax": 100,
      "setMin": 0,
      "rangeSegments": [0, 25, 50, 75, 100],
      "meterType": "vertical",
      "xAxis": ["Performance"]
    },
    width = 600,
    height = 600
  )
)
```
