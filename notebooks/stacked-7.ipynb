# Long sorted stacked bars with four components

A long sorted list of building materials, each bar split into four life-cycle stages, with the totals skewed towards a few items. Horizontal Stacked bar graph with a top legend and value labels.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/stacked-7.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).
