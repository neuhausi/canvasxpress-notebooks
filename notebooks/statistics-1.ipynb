# One comparison: raw p-value bracket

The simplest case: a single Welch t-test between two PlantGrowth treatments drawn as a bracket labelled with the raw p-value. The bracket height is set by hand.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/statistics-1.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "statistics1",
    data = {
      "y": {
        "vars": ["weight"],
        "smps": ["Obs1", "Obs2", "Obs3", "Obs4", "Obs5", "Obs6", "Obs7", "Obs8", "Obs9", "Obs10", "Obs11", "Obs12", "Obs13", "Obs14", "Obs15", "Obs16", "Obs17", "Obs18", "Obs19", "Obs20", "Obs21", "Obs22", "Obs23", "Obs24", "Obs25", "Obs26", "Obs27", "Obs28", "Obs29", "Obs30"],
        "data": [
          [4.17, 5.58, 5.18, 6.11, 4.5, 4.61, 5.17, 4.53, 5.33, 5.14, 4.81, 4.17, 4.41, 3.59, 5.87, 3.83, 6.03, 4.89, 4.32, 4.69, 6.31, 5.12, 5.54, 5.5, 5.37, 5.29, 4.92, 6.15, 5.8, 5.26]
        ]
      },
      "x": {
        "group": ["ctrl", "ctrl", "ctrl", "ctrl", "ctrl", "ctrl", "ctrl", "ctrl", "ctrl", "ctrl", "trt1", "trt1", "trt1", "trt1", "trt1", "trt1", "trt1", "trt1", "trt1", "trt1", "trt2", "trt2", "trt2", "trt2", "trt2", "trt2", "trt2", "trt2", "trt2", "trt2"]
      }
    },
    config = {
      "graphType": "Boxplot",
      "graphOrientation": "vertical",
      "groupingFactors": ["group"],
      "xAxis": ["weight"],
      "xAxisTitle": "weight",
      "smpTitle": "group",
      "smpTitleFontStyle": "bold",
      "axisTitleFontStyle": "bold",
      "axisAlgorithm": "rPretty",
      "smpTextRotate": 0,
      "showLegend": False,
      "colorBy": "group",
      "title": "PlantGrowth: dried weight by treatment (Welch t-test)",
      "showStatInference": True,
      "statInference": {
        "annotate": "p",
        "comparisons": [
          {
            "groups": ["trt1", "trt2"],
            "p": 0.00929840471726985,
            "value": 7.0
          }
        ]
      },
      "showBoxplotOriginalData": True,
      "jitter": True,
      "colorScheme": "Tableau"
    },
    width = 600,
    height = 600
  )
)
```
