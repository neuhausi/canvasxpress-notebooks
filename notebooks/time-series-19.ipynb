# Yearly totals as percent of the mean, two panels

Twenty-five yearly totals of rainfall and sunshine shown as percent of the 25-year mean in two stacked column panels, with recent years in an accent colour. Bar graph segregated into panels by a Panel annotation and coloured by a Period annotation.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/time-series-19.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "timeseries19",
    data = {
      "y": {
        "vars": ["Rainfall (% of mean)", "Sunshine (% of mean)"],
        "smps": ["2000 ", "2001 ", "2002 ", "2003 ", "2004 ", "2005 ", "2006 ", "2007 ", "2008 ", "2009 ", "2010 ", "2011 ", "2012 ", "2013 ", "2014 ", "2015 ", "2016 ", "2017 ", "2018 ", "2019 ", "2020 ", "2021 ", "2022 ", "2023 ", "2024 "],
        "data": [
          [97, 91, 99, 115, 98, 88, 105, 84, 88, 113, 100, 102, 84, 112, 106, 87, 95, 105, 96, 103, 93, 135, 124, 117, 102],
          [98, 96, 114, 94, 110, 94, 94, 115, 102, 84, 90, 99, 106, 114, 103, 94, 93, 97, 115, 109, 113, 111, 71, 68, 92]
        ]
      },
      "x": {
        "Period": ["Earlier", "Earlier", "Earlier", "Earlier", "Earlier", "Earlier", "Earlier", "Earlier", "Earlier", "Earlier", "Earlier", "Earlier", "Earlier", "Earlier", "Earlier", "Earlier", "Earlier", "Earlier", "Earlier", "Earlier", "Earlier", "Earlier", "Earlier", "Recent", "Recent"]
      },
      "z": {
        "Panel": ["Rainfall", "Sunshine"]
      }
    },
    config = {
      "graphType": "Bar",
      "theme": "ggplot",
      "titleAlign": "plot::0",
      "subtitleAlign": "plot::0",
      "graphOrientation": "vertical",
      "xAxis": ["Rainfall (% of mean)", "Sunshine (% of mean)"],
      "segregateVariablesBy": ["Panel"],
      "colorBy": "Period",
      "colorScheme": "User",
      "colors": ["#9ec9d9", "#0072B2"],
      "showLegend": False,
      "smpLabelInterval": 5,
      "smpTextRotate": 0,
      "xAxisTitle": "% of 25-year mean",
      "stripBackgroundColor": "rgba(255,255,255,0)",
      "stripBackgroundBorderColor": "rgba(255,255,255,0)",
      "stripTextAlign": "left",
      "stripTextFontStyle": "bold",
      "decorations": {
        "line": [
          {
            "color": "#444444",
            "value": 100,
            "width": 1.5,
            "type": "dashed",
            "label": "25-YEAR AVERAGE",
            "align": "left",
            "position": "under"
          }
        ],
        "text": [
          {
            "catX": 16.5,
            "value": 28,
            "label": "Wettest years\nin a while",
            "panel": 1,
            "color": "#0072B2",
            "labelScaleFontFactor": 1.15
          },
          {
            "catX": 16,
            "value": 28,
            "label": "2024 was\nparticularly cloudy",
            "panel": 2,
            "color": "#0072B2",
            "labelScaleFontFactor": 1.15
          }
        ]
      },
      "title": "Weather at a synthetic station is usually quite predictable",
      "subtitle": "Yearly totals as % of the 25-year average, 2000-2024 (synthetic)"
    },
    width = 760,
    height = 520
  )
)
```
