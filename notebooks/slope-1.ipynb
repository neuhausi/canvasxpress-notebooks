# Slope Chart

Slope charts connect each series between two time points or conditions, making the direction and magnitude of change immediately legible — who rose, who fell, and by how much — with the value labelled at each end.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/slope-1.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "slope1",
    data = {
      "y": {
        "vars": ["North", "South", "East", "West", "Central"],
        "smps": ["2015", "2024"],
        "data": [
          [42, 58],
          [61, 49],
          [35, 67],
          [54, 52],
          [28, 71]
        ]
      }
    },
    config = {
      "graphType": "Slope",
      "title": "Slope Chart",
      "xAxisTitle": "Year",
      "yAxisTitle": "Market Share (%)"
    },
    width = 600,
    height = 600
  )
)
```
