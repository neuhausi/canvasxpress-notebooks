# Min-max range plot on a log axis

Eight species as horizontal ranges from the lowest to the highest audible frequency, spanning tens of hertz to over 100 kHz, with the end values printed on the bars. Dumbbell graph with value labels.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/dumbbell-12.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "dumbbell12",
    data = {
      "y": {
        "vars": ["Lowest audible (kHz)", "Highest audible (kHz)"],
        "smps": ["Human", "Dog", "Elephant", "Bat", "Dolphin", "Moth", "Mouse", "Cat"],
        "data": [
          [0.024, 0.018, 0.006, 1, 0.18, 8, 2.3, 0.048],
          [19, 44, 11, 110, 140, 240, 92, 85]
        ]
      }
    },
    config = {
      "graphType": "Dumbbell",
      "dumbbellType": "connected",
      "theme": "stata",
      "colorScheme": "User",
      "colors": ["#0072B2", "#0072B2"],
      "title": "Hearing range of eight species",
      "subtitle": "Lowest and highest frequency each animal can hear, in kHz",
      "xAxis": ["Lowest audible (kHz)", "Highest audible (kHz)"],
      "xAxisTitle": "Frequency (kHz)",
      "setMinX": 0,
      "setMaxX": 260,
      "sortDir": "ascending",
      "showLegend": False,
      "dataPointSizeScaleFactor": 2.2,
      "citation": "Source: synthetic data for illustration"
    },
    width = 760,
    height = 460
  )
)
```
