# Color-coded Bar Graph for Continuous Data

Utilizing colors in a bar graph to represent data points with a continuous category.

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/bar-8.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "bar8",
    data = {
      "x": {
        "Drug Sensitivity": ["Sensitive", "Sensitive", "Sensitive", "Sensitive", "Sensitive", "Resistant", "Resistant", "Resistant"],
        "IC50": [0.08, 0.07, 0.06, 0.05, 0.04, 0.03, 0.02, 0.01]
      },
      "y": {
        "data": [
          [10, 15, 20, 30, 40, 70, 80, 90]
        ],
        "desc": ["Intensity"],
        "smps": ["S1", "S2", "S3", "S4", "S5", "S6", "S7", "S8"],
        "vars": ["V1"]
      }
    },
    config = {
      "colorBy": "IC50",
      "decorations": {
        "line": [
          {
            "align": "left",
            "color": "rgb(255,0,0)",
            "label": "Cutoff",
            "value": 50,
            "width": 2
          }
        ]
      },
      "graphOrientation": "vertical",
      "graphType": "Bar",
      "smpOverlays": ["Drug Sensitivity"],
      "smpTextRotate": 90,
      "smpTitle": "Cell Lines",
      "title": "Sensitivity of cell lines to different drugs",
      "xAxis": ["V1"]
    },
    width = 600,
    height = 600
  )
)
```
