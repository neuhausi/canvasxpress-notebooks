# Tumor evolution in breast cancer pre & post neoadjuvant treatment.

Visualizing tumor evolution in breast cancer pre- and post-neoadjuvant aromatase inhibitor therapy.

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/fish-3.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "fish3",
    data = {
      "y": {
        "data": [
          [100, 100],
          [98, 44],
          [46, 1],
          [1, 55]
        ],
        "smps": ["Primary", "Post-AI"],
        "vars": ["S1", "S2", "S3", "S4"]
      }
    },
    config = {
      "backgroundType": "panelSolidGradient",
      "colorSpectrum": ["#ffe4c4", "#ffb90f", "#cd6600"],
      "colors": ["#1B9E77", "#D95F02", "#7570B3", "#E7298A"],
      "fishAxis": [0, 120],
      "fishAxisLabels": ["Primary", "Post-AI"],
      "fishParents": [0, 1, 2, 1],
      "fishShape": "polygon",
      "fishTimepoints": [0, 120],
      "gradientOrientation": "horizontal",
      "graphType": "Fish",
      "legendKeyBackgroundBorderColor": "rgba(255,255,255,0)",
      "legendKeyBackgroundColor": "rgba(255,255,255,0)",
      "title": "BRC32",
      "xAxis": ["S1", "S2", "S3", "S4"]
    },
    width = 600,
    height = 450
  )
)
```
