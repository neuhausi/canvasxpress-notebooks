# Scatter Plot with Compact Legend Bins

Live example, data and code in other languages: https://www.canvasxpress.org/examples/scatter2d-24.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "scatter2d24",
    data = {
      "y": {
        "vars": ["USA", "Germany", "Japan", "Brazil", "India", "Nigeria", "China", "Indonesia"],
        "smps": ["GDP per capita", "Life expectancy"],
        "data": [
          [65000, 79],
          [53000, 81],
          [42000, 84],
          [9000, 76],
          [2100, 70],
          [2200, 55],
          [10500, 77],
          [4100, 72]
        ]
      },
      "z": {
        "Population": [331000000, 83000000, 126000000, 213000000, 1380000000, 206000000, 1440000000, 273000000]
      }
    },
    config = {
      "graphType": "Scatter2D",
      "xAxis": ["GDP per capita"],
      "yAxis": ["Life expectancy"],
      "sizeBy": "Population",
      "title": "Income vs life expectancy",
      "xAxisNumberFormat": {
        "style": "currency",
        "currency": "USD",
        "decimals": 0
      },
      "legendNumberFormat": "compact",
      "tooltipNumberFormat": "comma",
      "citation": "legendNumberFormat: compact population bins"
    },
    width = 600,
    height = 600
  )
)
```
