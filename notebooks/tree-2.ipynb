# Efficient Tree Graphs for Data Hierarchy

A tree graph displays hierarchial data, with nodes branching from a root node, enabling efficient search and sort algorithms.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/tree-2.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "tree2",
    data = {
      "x": {
        "Level1": [None, "L1", "L1", "L1", None],
        "Level2": [None, None, "L2", "L2", None]
      },
      "y": {
        "data": [
          [1, 2, 3, 4, 5]
        ],
        "smps": ["A", "B", "C", "D", "E"],
        "vars": ["Order"]
      }
    },
    config = {
      "graphType": "Tree",
      "hierarchy": ["Level1", "Level2"],
      "showTransition": True,
      "title": "Collapsible Tree",
      "xAxis": ["Order"]
    },
    width = 600,
    height = 600
  )
)
```
