# Area Line Graphs for Trend Analysis

Area line graphs show trends over time with emphasis on cumulative effect, ideal for financial performance or website traffic analysis.

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/arealine-3.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "arealine3",
    data = {
      "y": {
        "data": [
          [10, 11, 13, 4, 18, 21],
          [5, 8, 10, 12, 15, 10],
          [2, 3, 6, 4, 8, 6],
          [18, 27, 29, 21, 42, 37],
          [6, 9, 9, 6, 14, 12]
        ],
        "smps": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
        "vars": ["Series A", "Series B", "Series C", "Series D", "Series E"]
      }
    },
    config = {
      "colors": ["rgb(160,81,149)", "rgb(47,75,124)", "rgb(102,81,145)", "rgb(255,166,0)", "rgb(0,63,92)", "rgb(160,81,149)"],
      "graphOrientation": "vertical",
      "graphType": "AreaLine",
      "legendPosition": "topRight",
      "lineThickness": 3,
      "lineType": "spline",
      "setMinX": 0,
      "setMinX2": 0,
      "smpTextRotate": 90,
      "smpTitle": "Month",
      "xAxis": ["Series A"],
      "xAxis2": ["Series D"],
      "xAxisShow": True,
      "xAxis2Show": True,
      "xAxisTitle": "Value",
      "xAxis2Title": "Value2"
    },
    width = 600,
    height = 600
  )
)
```
