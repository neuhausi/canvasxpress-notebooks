# TrendPlot (percent-stacked, scale_y)

A percent-stacked TrendPlot: each category is normalized to 100% (areaType:'percent', the plotthis scale_y = TRUE form) with solid bars over a semi-transparent stacked area. areaShowColumns adds the bars; areaColumnWidth sets their width.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/area-20.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "area20",
    data = {
      "y": {
        "vars": ["Chrome", "Safari", "Edge", "Firefox"],
        "smps": ["FY21", "FY22", "FY23", "FY24", "FY25"],
        "data": [
          [61, 63, 64, 65, 66],
          [18, 19, 19, 20, 20],
          [8, 9, 10, 11, 12],
          [9, 7, 6, 5, 4]
        ]
      }
    },
    config = {
      "graphType": "Area",
      "graphOrientation": "vertical",
      "areaType": "percent",
      "areaStyle": "translucent",
      "areaShowColumns": True,
      "areaColumnWidth": 0.62,
      "colorScheme": "Tableau",
      "title": "TrendPlot: browser market share",
      "subtitle": "Percent-stacked bars with a stacked area trend (scale_y = TRUE)",
      "smpTitle": "Year",
      "xAxisTitle": "Share (%)",
      "xAxis": ["Chrome", "Safari", "Edge", "Firefox"]
    },
    width = 600,
    height = 600
  )
)
```
