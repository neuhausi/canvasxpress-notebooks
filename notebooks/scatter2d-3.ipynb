# Interactive Scatter2D Plot with Loess Fit Line

Interactive Scatter2D graph displaying data points with a loess fit line and confidence intervals.

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/scatter2d-3.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "scatter2d3",
    data = {
      "y": {
        "data": [
          [0.665, 1.561],
          [0.701, 1.99],
          [0.71, 2.118],
          [0.767, 3.834],
          [0.801, 4.602],
          [0.807, 5.199],
          [0.825, 4.255],
          [0.831, 4.818],
          [0.891, 5.064],
          [0.902, 5.283],
          [0.928, 5.344],
          [0.97, 4.691],
          [0.973, 5.055],
          [0.98, 4.937],
          [0.997, 3.752],
          [1, 3.965],
          [1.021, 3.275],
          [1.045, 2.849],
          [1.074, 2.286],
          [1.089, 1.64],
          [1.148, 0.97],
          [1.224, 0.537]
        ],
        "smps": ["E", "NOx"],
        "vars": ["V1", "V2", "V3", "V4", "V5", "V6", "V7", "V8", "V9", "V10", "V11", "V12", "V13", "V14", "V15", "V16", "V17", "V18", "V19", "V20", "V21", "V22"]
      }
    },
    config = {
      "graphType": "Scatter2D",
      "showConfidenceIntervals": True,
      "showLoessFit": True,
      "title": "Loess Fit",
      "xAxis": ["E"],
      "yAxis": ["NOx"]
    },
    width = 600,
    height = 600
  )
)
```
