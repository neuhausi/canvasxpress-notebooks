# Efficient Two-Series Data Comparison Bar Graph

A simple bar graph with two series showcasing data comparison efficiently.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/bar-2.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "bar2",
    data = {
      "y": {
        "data": [
          [4, 5, 4, 4, 7],
          [3, 6, 5, 5, 6]
        ],
        "smps": ["Cat 1", "Cat 2", "Cat 3", "Cat 4", "Cat 5"],
        "vars": ["Var 1", "Var 2"]
      }
    },
    config = {
      "dataTextColor": "#FFFFFF",
      "dataValuesPosition": "inside",
      "graphOrientation": "horizontal",
      "graphType": "Bar",
      "showDataValues": True,
      "smpTextRotate": 90,
      "smpTitle": "Categories",
      "title": "Bar graph showing data values",
      "xAxis": ["Var 1", "Var 2"],
      "xAxisTitle": "Value"
    },
    width = 600,
    height = 600
  )
)
```
