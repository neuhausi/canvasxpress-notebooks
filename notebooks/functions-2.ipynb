# 2D Function Scatter Plot

A 2D function scatter plot shows points with different y-coordinates.

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/functions-2.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "functions2",
    data = {
      "fx": "x^2"
    },
    config = {
      "graphType": "Scatter2D"
    },
    width = 600,
    height = 600
  )
)
```
