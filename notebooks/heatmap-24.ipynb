# Confusion matrix (model evaluation)

Confusion matrix of a 3-class classifier rendered with CanvasXpress.confusionConfig — counts shown per cell.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/heatmap-24.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "heatmap24",
    data = {
      "y": {
        "vars": ["Actual setosa", "Actual versicolor", "Actual virginica"],
        "smps": ["Pred setosa", "Pred versicolor", "Pred virginica"],
        "data": [
          [50, 0, 0],
          [0, 47, 3],
          [0, 5, 45]
        ]
      }
    },
    config = {
      "graphType": "Heatmap",
      "title": "Confusion matrix — iris classifier",
      "varTitle": "Actual",
      "smpTitle": "Predicted",
      "heatmapCellBox": True,
      "heatmapCellText": [
        {
          "variable": "Actual setosa",
          "sample": "Pred setosa",
          "text": "50"
        },
        {
          "variable": "Actual setosa",
          "sample": "Pred versicolor",
          "text": "0"
        },
        {
          "variable": "Actual setosa",
          "sample": "Pred virginica",
          "text": "0"
        },
        {
          "variable": "Actual versicolor",
          "sample": "Pred setosa",
          "text": "0"
        },
        {
          "variable": "Actual versicolor",
          "sample": "Pred versicolor",
          "text": "47"
        },
        {
          "variable": "Actual versicolor",
          "sample": "Pred virginica",
          "text": "3"
        },
        {
          "variable": "Actual virginica",
          "sample": "Pred setosa",
          "text": "0"
        },
        {
          "variable": "Actual virginica",
          "sample": "Pred versicolor",
          "text": "5"
        },
        {
          "variable": "Actual virginica",
          "sample": "Pred virginica",
          "text": "45"
        }
      ],
      "showTransition": False,
      "colorSpectrum": ["#f7fbff", "#6baed6", "#08306b"],
      "showHeatmapIndicator": True
    },
    width = 680,
    height = 600
  )
)
```
