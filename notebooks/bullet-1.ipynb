# Bullet Graph: Performance at a Glance

Bullet graph displaying data value within colored qualitative ranges and including a target marker for quick performance comparison.

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/bullet-1.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "bullet1",
    data = {
      "y": {
        "data": [
          [19, 25, 52, 57, 31, 42],
          [15, 15, 15, 20, 20, 20],
          [35, 35, 35, 40, 40, 40],
          [70, 70, 70, 70, 70, 70],
          [46, 52, 42, 63, 56, 51]
        ],
        "smps": ["S1", "S2", "S3", "S4", "S5", "S6"],
        "vars": ["V1", "V2", "V3", "V4", "V5"]
      }
    },
    config = {
      "rangeColors": ["#777777", "#AAAAAA", "#DDDDDD"],
      "rangeStack": ["V2", "V3", "V4"],
      "bulletTargetVarName": "V5",
      "dataTextColor": "#000000",
      "dataValuesPosition": "inside",
      "graphOrientation": "horizontal",
      "graphType": "Bullet",
      "showDataValues": True,
      "showLegend": False,
      "xAxis": ["V1"]
    },
    width = 600,
    height = 600
  )
)
```
