# 3D Scatter Plot Analysis of Math Functions

Visualize mathematical functions in a 3D scatter plot to analyze and understand their complex behaviors.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/functions-4.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "functions4",
    data = {
      "fx": "5 * sin(sqrt(x*x+y*y)) / sqrt(x*x+y*y)"
    },
    config = {
      "graphType": "Scatter3D",
      "setMaxX": 5,
      "setMaxY": 5,
      "setMaxZ": 5,
      "setMinX": -5,
      "setMinY": -5,
      "setMinZ": -5,
      "show3DGrid": False
    },
    width = 600,
    height = 600
  )
)
```
