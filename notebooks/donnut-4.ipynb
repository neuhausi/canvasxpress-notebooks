# Two donuts comparing composition by mass

Earth's crust and the Sun as two donuts on the same six element categories, so the contrast in composition is immediate, with a shared legend. Pie graph with pieInnerRadius and two samples.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/donnut-4.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "donnut4",
    data = {
      "y": {
        "vars": ["Oxygen", "Silicon", "Hydrogen", "Helium", "Iron", "Other"],
        "smps": ["Earth crust", "Sun"],
        "data": [
          [47, 1],
          [28, 1],
          [1, 73],
          [0.2, 25],
          [5, 1],
          [19, 0.2]
        ]
      }
    },
    config = {
      "graphType": "Pie",
      "theme": "paulTol",
      "pieInnerRadius": 0.5,
      "pieSegmentLabels": "inside",
      "pieSegmentPrecision": 0,
      "pieSegmentSeparation": 0,
      "showPieGrid": False,
      "showPieSampleLabel": True,
      "colorScheme": "User",
      "colors": ["#4477AA", "#66CCEE", "#228833", "#CCBB44", "#EE6677", "#BBBBBB"],
      "xAxis": ["Earth crust", "Sun"],
      "title": "The Sun is hydrogen; the crust is oxygen and silicon",
      "subtitle": "Share of mass by element, %, synthetic round numbers",
      "legendPosition": "left",
      "showTransition": False
    },
    width = 760,
    height = 520
  )
)
```
