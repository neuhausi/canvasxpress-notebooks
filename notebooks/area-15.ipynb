# Visualizing Trends: Non-Uniform Area Graph

Area graph with non-uniform data points across two series showcases trends effectively, emphasizing cumulative values for each category.

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/area-15.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "area15",
    data = {
      "y": {
        "data": [
          [1, 2],
          [3, 5],
          [5, 1],
          [2, 3],
          [4, 6],
          [6, 7]
        ],
        "smps": ["x", "y"],
        "vars": ["1", "2", "3", "4", "5", "6"]
      },
      "z": {
        "g": ["a", "a", "a", "b", "b", "b"]
      }
    },
    config = {
      "colorBy": "g",
      "colorScheme": "GGPlot",
      "graphType": "Area",
      "theme": "GGPlot",
      "xAxis": ["x"],
      "yAxis": ["y"]
    },
    width = 600,
    height = 600
  )
)
```
