# Market Trade Volume by Region

Treemap in the SCImago Graphica catalogue style: countries sized by market trade volume, grouped by region with colored group headers and a flat pastel palette per region.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/treemap-5.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "treemap5",
    data = {
      "y": {
        "vars": ["Trade"],
        "smps": ["India", "China", "Japan", "Iran", "Israel", "Pakistan", "Laos", "Mongolia", "France", "Germany", "Sweden", "UK", "Italy", "USA", "Mexico", "Canada", "Brazil", "S. Africa", "Egypt", "Sudan", "Congo", "Zaire"],
        "data": [
          [40, 36, 20, 18, 12, 11, 4, 1, 42, 31, 22, 21, 17, 52, 24, 16, 11, 30, 21, 12, 10, 8]
        ]
      },
      "x": {
        "Region": ["Asia", "Asia", "Asia", "Asia", "Asia", "Asia", "Asia", "Asia", "Europe", "Europe", "Europe", "Europe", "Europe", "America", "America", "America", "America", "Africa", "Africa", "Africa", "Africa", "Africa"],
        "Change": [63, 4, -12, 13, 24, -52, 3, 5, -11, -2, 13, -5, 4, 31, 12, -23, 10, 43, 0, 2, 12, 10]
      }
    },
    config = {
      "graphType": "Treemap",
      "groupingFactors": ["Region"],
      "xAxis": ["Trade"],
      "colorBy": "Region",
      "colorKey": {
        "Region": {
          "Asia": "#a2d46e",
          "Europe": "#a9c8e6",
          "America": "#f59c8d",
          "Africa": "#c79bd3"
        }
      },
      "treemapBorderColor": "rgb(255,255,255)",
      "treemapBorderWidth": 4,
      "showLegend": False,
      "title": "Market trade volume",
      "subtitle": "Countries sized by trade volume, grouped and colored by region"
    },
    width = 600,
    height = 600
  )
)
```
