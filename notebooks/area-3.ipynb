# Overlap Area Graph with Spline

An area graph illustrating three overlapping data series in a wide format, connected by a spline line for clear visualization.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/area-3.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "area3",
    data = {
      "y": {
        "data": [
          [10, 11, 13, 4, 18, 21],
          [5, 6, 10, 12, 15, 10],
          [2, 3, 6, 4, 8, 6]
        ],
        "smps": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
        "vars": ["Series A", "Series B", "Series C"]
      }
    },
    config = {
      "colorScheme": "Behance",
      "graphOrientation": "vertical",
      "graphType": "Area",
      "lineType": "spline",
      "objectColorTransparency": 0.7,
      "smpTextRotate": 90,
      "smpTitle": "Month / First Quarters of 2024",
      "subtitle": "random data",
      "title": "Area graph with three overlapping data series",
      "titleScaleFontFactor": 1.2,
      "xAxis": ["Series A", "Series B", "Series C"],
      "xAxisTitle": "Revenue (in Millions)"
    },
    width = 600,
    height = 600
  )
)
```
