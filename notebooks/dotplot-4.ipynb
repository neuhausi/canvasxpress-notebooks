# Master Data Visualization with Dot Plots

Dot plots are ideal for visualizing data distributions, offering a clear and intuitive way to show the frequency of individual data points.

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/dotplot-4.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "dotplot4",
    data = {
      "y": {
        "data": [
          [94, 96, 112, 92, 90, 78, 94, 76, 79, 86, 93, 84, 67, 73, 80, 62, 72, 71, 68, 64, 72],
          [152, 151, 165, 141, 137, 118, 131, 112, 114, 119, 124, 114, 94, 100, 107, 84, 92, 88, 82, 78, 81]
        ],
        "smps": ["MIT", "Stanford", "Harvard", "U.Penn", "Princeton", "Chicago", "Georgetown", "Tufts", "Yale", "Columbia", "Duke", "Dartmouth", "NYU", "Notre Dame", "Cornell", "Michigan", "Brown", "Berkeley", "Emory", "UCLA", "SoCal"],
        "vars": ["Women", "Men"]
      }
    },
    config = {
      "axisAlgorithm": "wilkinson",
      "dataPointSizeScaleFactor": 3,
      "dataTextScaleFontFactor": 0.6,
      "dotplotType": "stacked",
      "graphType": "Dotplot",
      "plotStyle": "open",
      "showDataValues": True,
      "smpTitle": "School",
      "sortDir": "descending",
      "title": "Gender Earnings Disparity",
      "xAxis": ["Women", "Men"],
      "xAxis2Title": "Annual Salary",
      "xAxisGridMinorShow": False,
      "xAxisTickFormat": "$%sK",
      "xAxisTitle": "Annual Salary"
    },
    after_render = [
      [
        "sortSamplesByVariable",
        ["Men"]
      ]
    ],
    width = 600,
    height = 600
  )
)
```
