# Sub-national regions choropleth

Namibia's regions coloured by annual rainfall on a sequential blue ramp, with region labels and a legend. Map on a country-level topoJSON matched by feature order.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/map-23.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "map23",
    data = {
      "y": {
        "vars": ["Karas", "Kunene", "Omusati", "Ohangwena", "Kavango", "Caprivi", "Hardap", "Otjozondjupa", "Omaheke", "Erongo", "Khomas", "Oshana", "Oshikoto"],
        "smps": ["Rainfall (mm)"],
        "data": [
          [70],
          [120],
          [330],
          [410],
          [520],
          [660],
          [160],
          [440],
          [360],
          [60],
          [300],
          [350],
          [400]
        ]
      }
    },
    config = {
      "graphType": "Map",
      "topoJSON": "https://www.canvasxpress.org/data/maps/NAM.json",
      "mapId": "namibia",
      "mapConfig": {
        "zoom": 5.6,
        "center": [-22.9, 18.5]
      },
      "theme": "bw",
      "colorBy": "Rainfall (mm)",
      "colorScheme": "User",
      "colorSpectrum": ["#dfc27d", "#a6b37a", "#35978f", "#01665e"],
      "showMapLabels": True,
      "title": "Annual rainfall per region",
      "subtitle": "Regional mean of total yearly rainfall in mm (synthetic)",
      "legendPosition": "topRight",
      "showLegendTitle": True,
      "mapOutlineColor": "#ffffff",
      "source": "Synthetic data for illustration"
    },
    width = 760,
    height = 600
  )
)
```
