# Arrow plot in grouped panels with shaded bands

Fifteen ecosystem classes in three panels, each row an arrow from the 2000s to the 2020s value, with a shaded band showing the typical range of each class. Dumbbell graph with arrows, segregated by a Group annotation.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/dumbbell-15.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "dumbbell15",
    data = {
      "y": {
        "vars": ["2000s", "2020s"],
        "smps": ["Kelp", "Eelgrass", "Mangrove", "Saltmarsh", "Seagrass", "Tundra", "Taiga", "Grassland", "Savanna", "Steppe", "Desert", "Chaparral", "Alpine", "Arctic", "Wetland"],
        "data": [
          [560, 610, 590, 575, 630, 548, 512, 538, 525, 500, 470, 420, 430, 405, 415],
          [545, 640, 568, 598, 605, 536, 532, 510, 505, 508, 414, 390, 452, 385, 438]
        ]
      },
      "x": {
        "Group": ["1. High productivity", "1. High productivity", "1. High productivity", "1. High productivity", "1. High productivity", "2. Moderate productivity", "2. Moderate productivity", "2. Moderate productivity", "2. Moderate productivity", "2. Moderate productivity", "3. Low productivity", "3. Low productivity", "3. Low productivity", "3. Low productivity", "3. Low productivity"]
      }
    },
    config = {
      "graphType": "Dumbbell",
      "dumbbellType": "arrow",
      "dumbbellArrowDirection": "order",
      "theme": "bw",
      "colorScheme": "User",
      "colors": ["#0072B2", "#0072B2"],
      "title": "Carbon-uptake index by ecosystem class",
      "subtitle": "Index of net carbon uptake per ecosystem, 2000s to 2020s; shaded: the typical band of each class",
      "xAxis": ["2000s", "2020s"],
      "xAxisTitle": "Uptake index",
      "connectByColor": "#0072B2",
      "connectByWidth": 3,
      "dataPointSizeScaleFactor": 0.8,
      "setMinX": 380,
      "setMaxX": 660,
      "smpLabelInterval": 1,
      "showLegend": False,
      "segregateSamplesBy": ["Group"],
      "stripBackgroundColor": "rgba(255,255,255,0)",
      "stripTextAlign": "left",
      "stripBackgroundBorderColor": "rgba(255,255,255,0)",
      "layoutTopology": "3X1",
      "decorations": {
        "range": [
          {
            "value": 540,
            "value2": 640,
            "color": "rgba(0,158,115,0.14)",
            "scope": "1. High productivity",
            "tooltip": "Typical high-productivity band"
          },
          {
            "value": 500,
            "value2": 550,
            "color": "rgba(230,159,0,0.16)",
            "scope": "2. Moderate productivity",
            "tooltip": "Typical moderate-productivity band"
          },
          {
            "value": 390,
            "value2": 470,
            "color": "rgba(213,94,0,0.14)",
            "scope": "3. Low productivity",
            "tooltip": "Typical low-productivity band"
          }
        ],
        "text": [
          {
            "value": 510,
            "smpId": 0,
            "panel": 1,
            "label": "Typical range ▶",
            "color": "#009E73",
            "labelScaleFontFactor": 0.9
          },
          {
            "value": 582,
            "smpId": 0,
            "panel": 2,
            "label": "◀ Typical range",
            "color": "#B07C00",
            "labelScaleFontFactor": 0.9
          },
          {
            "value": 502,
            "smpId": 0,
            "panel": 3,
            "label": "◀ Typical range",
            "color": "#D55E00",
            "labelScaleFontFactor": 0.9
          }
        ]
      },
      "citation": "Source: synthetic data for illustration"
    },
    width = 760,
    height = 720
  )
)
```
