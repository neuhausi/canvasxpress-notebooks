# Visualize Data Flow with Sankey Diagrams

Sankey diagrams visually represent flow magnitudes and relationships in a single level, simplifying complex data.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/sankey-1.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "sankey1",
    data = {
      "x": {
        "Source": ["A", "A", "A", "B", "B", "B"],
        "Target": ["X", "Y", "Z", "X", "Y", "Z"],
        "Weight": [1, 2, 3, 4, 5, 6]
      },
      "y": {
        "data": [
          [5, 7, 6, 2, 9, 4]
        ],
        "smps": ["smp1", "smp2", "smp3", "smp4", "smp5", "smp6"],
        "vars": ["Weight"]
      }
    },
    config = {
      "graphOrientation": "vertical",
      "graphType": "Sankey",
      "sankeyNodesColor": "rgba(20, 250, 50, 0.4)",
      "sankeySource": "Source",
      "sankeyTarget": "Target",
      "showTransition": False,
      "title": "Single Level Sankey",
      "xAxis": ["Weight"]
    },
    width = 600,
    height = 600
  )
)
```
