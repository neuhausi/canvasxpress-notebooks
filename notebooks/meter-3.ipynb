# Digital Meter Graph for Data Analysis

Track performance metrics over time with a digital meter graph for effective data analysis and decision-making.

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/meter-3.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "meter3",
    data = {
      "y": {
        "data": [
          [85]
        ],
        "smps": ["January"],
        "vars": ["Performance"]
      }
    },
    config = {
      "graphType": "Meter",
      "setMax": 100,
      "setMin": 0,
      "rangeSegments": [0, 25, 50, 75, 100],
      "meterType": "digital",
      "xAxis": ["Performance"]
    },
    width = 600,
    height = 600
  )
)
```
