# Spline Area Graph for Clear Trend Visualization 

A spline area graph effectively showcases trends over time with a single data series, enhancing clarity in data visualization.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/area-2.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "area2",
    data = {
      "y": {
        "data": [
          [10, 11, 13, 4, 18, 21]
        ],
        "smps": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
        "vars": ["Value"]
      }
    },
    config = {
      "colorScheme": "Prism",
      "graphOrientation": "vertical",
      "graphType": "Area",
      "lineType": "spline",
      "showLegend": False,
      "smpTextRotate": 90,
      "smpTitle": "Month / First Quarters of 2024",
      "title": "Area graph with one series - Spline",
      "titleScaleFontFactor": 1.2,
      "xAxis": ["Value"],
      "xAxisTitle": "Revenue (in Millions)"
    },
    width = 600,
    height = 600
  )
)
```
