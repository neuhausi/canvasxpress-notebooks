# Diverging bars: ten positive, ten negative

Ten warming and ten cooling ocean regions as horizontal bars diverging from zero, coloured by direction through an annotation, sorted by value and labelled with the value. Bar graph with colorBy.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/bar-21.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "bar21",
    data = {
      "y": {
        "vars": ["Trend"],
        "smps": ["Barents Sea", "Kara Sea", "Laptev Sea", "Svalbard shelf", "Chukchi Sea", "Bering Strait", "Beaufort Sea", "Baffin Bay", "Labrador Sea", "Irminger Sea", "Subpolar gyre", "Gulf Stream ext.", "Sargasso Sea", "Azores front", "Canary current"],
        "data": [
          [0.83, 0.81, 0.76, 0.71, 0.58, 0.58, 0.53, 0.51, -0.26, -0.29, -0.29, -0.3, -0.31, -0.34, -0.38]
        ]
      },
      "x": {
        "Direction": ["Warming faster", "Warming faster", "Warming faster", "Warming faster", "Warming faster", "Warming faster", "Warming faster", "Warming faster", "Warming slower", "Warming slower", "Warming slower", "Warming slower", "Warming slower", "Warming slower", "Warming slower"],
        "Trend": [0.83, 0.81, 0.76, 0.71, 0.58, 0.58, 0.53, 0.51, -0.26, -0.29, -0.29, -0.3, -0.31, -0.34, -0.38]
      }
    },
    config = {
      "graphType": "Bar",
      "titleAlign": "plot::0",
      "subtitleAlign": "plot::0",
      "legendHorizontalJustification": 0,
      "legendJustifyTo": "plot",
      "graphOrientation": "horizontal",
      "theme": "paulTol",
      "colorBy": "Trend",
      "colorSpectrum": ["#B5651D", "#E9B96E", "#F6D9A0", "#4FB3A0", "#1B7F74", "#003D3B"],
      "colorSpectrumZeroValue": 0,
      "xAxis": ["Trend"],
      "showDataValues": True,
      "dataValuesPosition": "inside",
      "dataTextColor": "#ffffff",
      "dataTextScaleFontFactor": 0.8,
      "showLegend": False,
      "title": "Ocean regions where warming trends diverge most",
      "subtitle": "Change in sea-surface temperature trend, °C per decade, 2000–2016 versus 1960–2000 (illustrative).",
      "smpTitle": "",
      "xAxisTitle": "",
      "sampleSpaceFactor": 0.25
    },
    width = 760,
    height = 600
  )
)
```
