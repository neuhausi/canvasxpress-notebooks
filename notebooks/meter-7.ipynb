# Meter Graphs for Effective Monitoring

Meter graphs visualize data points over time, aiding in system monitoring and decision-making with clear trends and patterns.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/meter-7.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "meter7",
    data = {
      "x": {
        "Quarter": ["Q1", "Q1", "Q1", "Q1", "Q1", "Q1", "Q2", "Q2", "Q2", "Q2", "Q2", "Q2"]
      },
      "y": {
        "data": [
          [85, 60, 29, 10, 90, 98, 93, 65, 89, 34, 82, 67],
          [1000, 5000, 4000, 2000, 7000, 9000, 8000, 3000, 2000, 6000, 5000, 10000]
        ],
        "smps": ["January", "February", "March", "April", "May", "June", "July", "August", "September", "October", "November", "December"],
        "vars": ["Performance", "Revenue"]
      },
      "z": {
        "type": ["Performance", "Revenue"]
      }
    },
    config = {
      "graphType": "Meter",
      "meterType": "number",
      "meterVar": "Revenue",
      "xAxis": ["Revenue"]
    },
    width = 600,
    height = 600
  )
)
```
