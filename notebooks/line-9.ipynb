# Learning curve (model evaluation)

Training vs validation score over training-set size from CanvasXpress.learningCurveConfig.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/line-9.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "line9",
    data = {
      "y": {
        "vars": ["Training score", "Validation score"],
        "smps": ["50", "100", "250", "500", "1000", "2500", "5000"],
        "data": [
          [0.98, 0.97, 0.955, 0.95, 0.945, 0.942, 0.94],
          [0.82, 0.86, 0.89, 0.905, 0.915, 0.92, 0.922]
        ]
      }
    },
    config = {
      "graphType": "Line",
      "title": "Learning curve — training vs validation",
      "smpTitle": "Training set size",
      "yAxisTitle": "Score",
      "showLegend": True,
      "showTransition": False
    },
    width = 600,
    height = 600
  )
)
```
