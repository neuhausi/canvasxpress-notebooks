# Centered 2D Function Scatter Plot

Scatter plot displaying a 2D function centered at (0,0).

Live example, data and code in other languages: https://www.canvasxpress.org/examples/functions-3.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "functions3",
    data = {
      "fx": "1/x * cos(1/x)"
    },
    config = {
      "graphType": "Scatter2D"
    },
    width = 600,
    height = 600
  )
)
```
