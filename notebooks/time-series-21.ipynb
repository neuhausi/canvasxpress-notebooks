# Editorial slope chart: two years, one highlighted series

A slope chart of median income in 2014 and 2024 for five economies: a two-sample Line graph with end labels on both sides, one series highlighted in colour and the others in grey, on the editorial theme with a structured footer.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/time-series-21.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "timeseries21",
    data = {
      "y": {
        "vars": ["Canada", "United States", "United Kingdom", "France", "Japan"],
        "smps": ["2014 ", "2024 "],
        "data": [
          [36.1, 40.4],
          [39.5, 44.1],
          [30.2, 31.9],
          [31.3, 33],
          [27.9, 27.1]
        ]
      }
    },
    config = {
      "graphType": "Line",
      "graphOrientation": "vertical",
      "theme": "editorial",
      "colorScheme": "User",
      "title": "Japan is the only G7 economy where median income fell",
      "subtitle": "Equivalised median disposable income, thousand USD (PPP)",
      "showLegend": False,
      "showLineEndLabels": True,
      "dataTextScaleFontFactor": 0.8,
      "numberFormat": {
        "style": "decimal",
        "decimals": 0
      },
      "colors": ["#bbbbbb", "#bbbbbb", "#bbbbbb", "#bbbbbb", "#ee6677"],
      "yAxisTitle": "",
      "xAxisTitle": "",
      "source": "OECD",
      "byline": "Chart: CanvasXpress",
      "showDataDownload": True,
      "responsiveRules": [
        {
          "maxWidth": 480,
          "config": {
            "legendPosition": "bottom",
            "fontScaleFontFactor": 0.85
          }
        }
      ]
    },
    width = 760,
    height = 517
  )
)
```
