# Exponential Relationships Analysis for Industries

Visualize and analyze complex relationships with an exponential fit for accurate interpretation and prediction in various industries.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/nonlinear-fit-4.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "nonlinearfit4",
    data = {
      "y": {
        "data": [
          [0, 179.5],
          [5, 168.7],
          [8, 158.1],
          [11, 149.2],
          [15, 141.7],
          [18, 134.6],
          [22, 125.4],
          [25, 123.5],
          [30, 116.3],
          [34, 113.2],
          [38, 109.1],
          [42, 105.7],
          [45, 102.2],
          [50, 100.5]
        ],
        "smps": ["Time", "Temp"],
        "vars": ["S1", "S2", "S3", "S4", "S5", "S6", "S7", "S8", "S9", "S10", "S11", "S12", "S13", "S14"]
      }
    },
    config = {
      "decorations": {
        "exponential": [
          {
            "color": "red"
          }
        ]
      },
      "graphType": "Scatter2D",
      "showDecorations": True,
      "showLegend": False,
      "xAxis": ["Time"],
      "yAxis": ["Temp"]
    },
    width = 600,
    height = 600
  )
)
```
