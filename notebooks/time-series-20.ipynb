# Editorial line chart with a callout and end labels

Four annual unemployment series on the editorial theme: series names at the line ends instead of a legend, a curved decorations.marker callout, a percent axis format and a full structured footer (notes, linked source, byline, 'Get the data'). Line graph with showLineEndLabels and responsiveRules.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/time-series-20.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "timeseries20",
    data = {
      "y": {
        "vars": ["Spain", "Italy", "France", "Germany"],
        "smps": ["2015 ", "2016 ", "2017 ", "2018 ", "2019 ", "2020 ", "2021 ", "2022 ", "2023 ", "2024 "],
        "data": [
          [22.1, 19.6, 17.2, 15.3, 14.1, 15.5, 14.8, 12.9, 12.1, 11.4],
          [11.9, 11.7, 11.2, 10.6, 10, 9.2, 9.5, 8.1, 7.7, 6.5],
          [10.4, 10.1, 9.4, 9, 8.4, 8, 7.9, 7.3, 7.3, 7.4],
          [4.6, 4.1, 3.8, 3.4, 3.2, 3.8, 3.6, 3.1, 3, 3.4]
        ]
      }
    },
    config = {
      "graphType": "Line",
      "graphOrientation": "vertical",
      "theme": "editorial",
      "colorScheme": "User",
      "title": "Unemployment keeps falling in southern Europe",
      "colors": ["#4477aa", "#ee6677", "#ccbb44", "#228833"],
      "subtitle": "Unemployment rate, % of labour force, annual average",
      "showLegend": False,
      "showLineEndLabels": True,
      "dataTextScaleFontFactor": 0.8,
      "numberFormat": {
        "style": "decimal",
        "decimals": 0,
        "suffix": "%"
      },
      "tooltipNumberFormat": {
        "style": "decimal",
        "decimals": 1,
        "suffix": "%"
      },
      "yAxisTitle": "",
      "xAxisTitle": "",
      "decorations": {
        "marker": [
          {
            "text": "Pandemic spike",
            "variable": "Spain",
            "sample": "2020 ",
            "type": "curved",
            "x": 0.62,
            "y": 0.18
          }
        ]
      },
      "notes": "Annual averages; 2024 provisional.",
      "source": "Eurostat",
      "sourceUrl": "https://ec.europa.eu/eurostat",
      "byline": "Chart: CanvasXpress",
      "showDataDownload": True,
      "responsiveRules": [
        {
          "maxWidth": 480,
          "config": {
            "legendPosition": "bottom",
            "fontScaleFontFactor": 0.85
          }
        }
      ]
    },
    width = 760,
    height = 517
  )
)
```
