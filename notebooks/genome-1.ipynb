# Advantages of Genome Graphs in Diagnostics

Genome graphs offer a comprehensive view of genetic variation within populations, enhancing disease diagnostics and personalized medicine.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/genome-1.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).
