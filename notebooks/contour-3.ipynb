# Enhance Data with Loess Smoothing

A typical 3-column dataset requires smoothing using loess.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/contour-3.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "contour3",
    data = {
      "y": {
        "data": [
          [0, 0, 10],
          [1, 0, 5.625],
          [2, 0, 2.5],
          [3, 0, 0.625],
          [4, 0, 0],
          [0, 1, 10.625],
          [1, 1, 6.25],
          [2, 1, 3.125],
          [3, 1, 1.25],
          [4, 1, 0.625],
          [0, 2, 12.5],
          [1, 2, 8.125],
          [2, 2, 5],
          [3, 2, 3.125],
          [4, 2, 2.5],
          [0, 3, 15.625],
          [1, 3, 11.25],
          [2, 3, 8.125],
          [3, 3, 6.25],
          [4, 3, 5.625],
          [0, 4, 20],
          [1, 4, 15.625],
          [2, 4, 12.5],
          [3, 4, 10.625],
          [4, 4, 10]
        ],
        "smps": ["s1", "s2", "s3"],
        "vars": ["v1", "v2", "v3", "v4", "v5", "v6", "v7", "v8", "v9", "v10", "v11", "v12", "v13", "v14", "v15", "v16", "v17", "v18", "v19", "v20", "v21", "v22", "v23", "v24", "v25"]
      }
    },
    config = {
      "contourFilled": True,
      "graphType": "ScatterBubble2D",
      "showContourLevel": True,
      "title": "Basic Contour Plot",
      "xAxis": ["s1"],
      "yAxis": ["s2"],
      "zAxis": ["s3"]
    },
    after_render = [
      [
        "createContour",
        []
      ]
    ],
    width = 600,
    height = 600
  )
)
```
