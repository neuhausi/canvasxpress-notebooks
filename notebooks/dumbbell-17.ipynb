# Editorial arrow plot, 2014 to 2024

Renewable share of electricity for seven countries as arrows from the 2014 to the 2024 value, running in sample order so every arrow points from before to after. Dumbbell graph with dumbbellType arrow and dumbbellArrowDirection order on the editorial theme.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/dumbbell-17.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "dumbbell17",
    data = {
      "y": {
        "vars": ["2014", "2024"],
        "smps": ["Denmark", "Germany", "United Kingdom", "Spain", "United States", "Japan", "India"],
        "data": [
          [48, 26, 18, 37, 13, 14, 15],
          [81, 52, 46, 55, 23, 25, 22]
        ]
      }
    },
    config = {
      "graphType": "Dumbbell",
      "dumbbellType": "arrow",
      "dumbbellArrowDirection": "order",
      "legendColumns": 2,
      "arrowHeadSize": 8,
      "xAxisTicks": 4,
      "smpTextScaleFontFactor": 1.3,
      "theme": "editorial",
      "colorScheme": "User",
      "title": "Renewables have doubled their share in a decade",
      "subtitle": "Renewable share of electricity generation, 2014 → 2024, %",
      "legendPosition": "bottom",
      "numberFormat": {
        "style": "decimal",
        "decimals": 0,
        "suffix": "%"
      },
      "colors": ["#bbbbbb", "#228833"],
      "xAxisTitle": "",
      "notes": "Wind, solar, hydro and bioenergy. 2024 preliminary.",
      "source": "Ember",
      "sourceUrl": "https://ember-climate.org",
      "byline": "Chart: CanvasXpress",
      "showDataDownload": True,
      "responsiveRules": [
        {
          "maxWidth": 480,
          "config": {
            "legendPosition": "bottom",
            "fontScaleFontFactor": 0.85,
            "arrowHeadSize": 5,
            "connectByWidth": 1
          }
        }
      ]
    },
    width = 760,
    height = 517
  )
)
```
