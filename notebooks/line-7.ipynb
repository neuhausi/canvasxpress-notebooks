# Annotated Line with Direct Labels

Live example, data and code in other languages: https://www.canvasxpress.org/examples/line-7.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "line7",
    data = {
      "y": {
        "vars": ["Germany", "France", "Italy", "Spain"],
        "smps": ["Y2016", "Y2017", "Y2018", "Y2019", "Y2020", "Y2021"],
        "data": [
          [41.2, 42.0, 42.9, 38.9, 40.5, 42.2],
          [36.5, 37.1, 37.8, 34.2, 35.9, 37.5],
          [28.9, 29.4, 30.1, 26.8, 28.1, 29.5],
          [24.1, 24.8, 25.5, 22.0, 23.6, 25.1]
        ]
      }
    },
    config = {
      "graphType": "Line",
      "graphOrientation": "vertical",
      "title": "GDP per capita (thousand USD)",
      "a11yDescription": "GDP per capita fell in all four countries in 2019 and had recovered by 2021. Germany stays highest (42.2 thousand USD in 2021) and Spain lowest (25.1).",
      "showLineEndLabels": True,
      "decorations": {
        "marker": [
          {
            "text": "COVID-19 drop",
            "variable": "Germany",
            "sample": "Y2019",
            "type": "curved",
            "x": 0.42,
            "y": 0.12
          }
        ]
      },
      "source": "Eurostat",
      "byline": "Chart: CanvasXpress"
    },
    width = 600,
    height = 600
  )
)
```
