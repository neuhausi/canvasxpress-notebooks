# Efficient Data Trend Analysis with Light Meter Graph

Monitor and analyze data trends efficiently with a light meter graph format.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/meter-4.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "meter4",
    data = {
      "y": {
        "data": [
          [85]
        ],
        "smps": ["January"],
        "vars": ["Performance"]
      }
    },
    config = {
      "graphType": "Meter",
      "meterType": "state",
      "xAxis": ["Performance"]
    },
    width = 600,
    height = 600
  )
)
```
