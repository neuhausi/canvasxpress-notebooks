# Comparing Continuous Variables with Dodged Bar Histograms

Histograms displaying multiple series with dodged bars provide a visual comparison of the distribution of continuous variables in each series.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/histogram-4.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "histogram4",
    data = {
      "y": {
        "data": [
          [6.47, 4.03],
          [6.13, 3.76],
          [6.19, 3.77],
          [4.89, 3.34],
          [5.63, 3.47],
          [4.52, 2.92],
          [5.89, 3.2],
          [4.79, 2.71],
          [3.53, 3.53],
          [6.08, 4.51],
          [4.02, 4.56]
        ],
        "smps": ["Alcohol", "Tobacco"],
        "vars": ["North", "Yorkshire", "Northeast", "East Midlands", "West Midlands", "East Anglia", "Southeast", "Southwest", "Wales", "Scotland", "Northern Ireland"]
      }
    },
    config = {
      "citation": "Moore, David S., and George P. McCabe (1989). Introduction to the Practice of Statistics, p. 179.",
      "graphType": "Scatter2D",
      "histogramBins": 5,
      "showHistogram": True,
      "title": "Average weekly household spending, in British pounds, on tobacco products\\nand alcoholic beverages for each of the 11 regions of Great Britain.",
      "xAxis": ["Tobacco", "Alcohol"],
      "xAxisTitle": "Pounds Spent",
      "yAxisTitle": "Frequency"
    },
    width = 600,
    height = 600
  )
)
```
