# Dumbbell Chart: Visual Data Comparison

Visualize data relationships using dumbbell graphs that highlight the difference between two points across categories.

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/dumbbell-6.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "dumbbell6",
    data = {
      "y": {
        "vars": ["Combined", "Male", "Female"],
        "smps": ["Monaco", "Japan", "Germany", "Italy", "Greece", "Hong Kong", "Austria", "Spain", "Netherlands", "Finland", "South Korea", "France", "Poland", "United Kingdom", "Russia", "Norway", "United States", "New Zealand", "China"],
        "data": [
          [53.1, 47.3, 47.1, 45.5, 44.5, 44.4, 44, 42.7, 42.6, 42.5, 41.8, 41.4, 40.7, 40.5, 39.6, 39.2, 38.1, 37.9, 37.4],
          [51.7, 46, 46, 44.4, 43.5, 43.5, 42.8, 41.5, 41.5, 40.9, 40.2, 39.6, 39, 39.3, 36.6, 38.4, 36.8, 37.1, 36.5],
          [54.5, 48.7, 48.2, 46.5, 45.6, 45, 45.1, 43.9, 43.6, 44.3, 43.4, 43.1, 42.4, 41.7, 42.5, 40, 39.4, 38.8, 38.4]
        ]
      }
    },
    config = {
      "colors": ["grey", "blue", "pink"],
      "dumbbellType": "arrow",
      "graphType": "Dumbbell",
      "legendColumns": 2,
      "legendPosition": "bottom",
      "sortDir": "ascending",
      "title": "Age Range by Gender",
      "xAxis": ["Female", "Male"],
      "xAxis2Show": True,
      "xAxisShow": False,
      "xAxisTitle2": "Age"
    },
    width = 600,
    height = 600
  )
)
```
