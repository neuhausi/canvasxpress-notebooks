# Gauge Meter Card

A gauge meter framed as a dashboard card (meterCard) filling the canvas with a consistent margin.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/meter-12.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "meter12",
    data = {
      "y": {
        "data": [
          [85]
        ],
        "smps": ["January"],
        "vars": ["Performance"]
      }
    },
    config = {
      "graphType": "Meter",
      "setMax": 100,
      "setMin": 0,
      "meterType": "gauge",
      "meterCard": True,
      "xAxis": ["Performance"]
    },
    width = 600,
    height = 420
  )
)
```
