# Population Pyramid

A population pyramid places two groups back-to-back across shared age bands — here male and female counts — so the age-sex structure of a population reads at a glance. It renders through the Tornado back-to-back bar layout.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/pyramid-1.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "pyramid1",
    data = {
      "y": {
        "vars": ["Male", "Female"],
        "smps": ["0-14", "15-29", "30-44", "45-59", "60-74", "75+"],
        "data": [
          [-320, -298, -265, -210, -140, -70],
          [305, 290, 262, 215, 152, 96]
        ]
      }
    },
    config = {
      "graphType": "Pyramid",
      "graphOrientation": "horizontal",
      "title": "Population Pyramid",
      "xAxisTitle": "Population (thousands)",
      "smpTitle": "Age band"
    },
    width = 600,
    height = 600
  )
)
```
