# Strip Plot

Strip plots show every individual observation as a jittered point within each category, revealing the full distribution, clusters and outliers without hiding raw data behind a summary.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/stripplot-1.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "stripplot1",
    data = {
      "y": {
        "vars": ["Score"],
        "smps": ["s1", "s2", "s3", "s4", "s5", "s6", "s7", "s8", "s9", "s10", "s11", "s12", "s13", "s14", "s15", "s16", "s17", "s18"],
        "data": [
          [82, 85, 79, 88, 91, 84, 77, 90, 86, 61, 65, 58, 70, 63, 67, 55, 72, 60]
        ]
      },
      "x": {
        "Group": ["Control", "Control", "Control", "Control", "Control", "Control", "Control", "Control", "Control", "Low Dose", "Low Dose", "Low Dose", "Low Dose", "Low Dose", "Low Dose", "Low Dose", "Low Dose", "Low Dose"]
      }
    },
    config = {
      "graphType": "StripPlot",
      "graphOrientation": "vertical",
      "groupingFactors": ["Group"],
      "colorBy": "Group",
      "title": "Strip Plot",
      "yAxisTitle": "Score"
    },
    width = 600,
    height = 600
  )
)
```
