# Election Donut (Half Pie)

Live example, data and code in other languages: https://www.canvasxpress.org/examples/pie-5.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "pie5",
    data = {
      "y": {
        "vars": ["SPD", "CDU/CSU", "Greens", "FDP", "AfD", "Left"],
        "smps": ["Seats"],
        "data": [
          [206],
          [197],
          [118],
          [92],
          [83],
          [39]
        ]
      }
    },
    config = {
      "graphType": "Pie",
      "pieShape": "half",
      "pieInnerRadius": 0.55,
      "pieLabelType": "value",
      "title": "Election result (seats)",
      "colors": ["#e3000f", "#1f1f1f", "#46962b", "#e6c700", "#009ee0", "#be3075"],
      "legendPosition": "bottom",
      "legendColumns": 3,
      "source": "Bundeswahlleiterin"
    },
    width = 600,
    height = 600
  )
)
```
