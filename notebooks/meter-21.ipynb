# Ring Meter KPI

A ring meter renders a single KPI as a thick colored circle around a lightly tinted disc, with the value centered and the title above — a clean dashboard-style summary tile.

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/meter-21.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "meter21",
    data = {
      "y": {
        "vars": ["Full-Stack Developer"],
        "smps": ["Applicants"],
        "data": [
          [150]
        ]
      }
    },
    config = {
      "graphType": "Meter",
      "meterType": "ring",
      "summaryType": "sum",
      "meterCard": True,
      "xAxis": ["Full-Stack Developer"],
      "rangeColors": ["rgb(124,77,255)"]
    },
    width = 600,
    height = 600
  )
)
```
