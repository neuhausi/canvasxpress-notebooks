# Visual Data Comparison with Lollipop Graph

Eye-catching lollipop graph perfect for comparing data visually.

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/lollipop-1.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "lollipop1",
    data = {
      "y": {
        "data": [
          [10, 15, 20, 30, 40, 70, 80, 90]
        ],
        "smps": ["S1", "S2", "S3", "S4", "S5", "S6", "S7", "S8"],
        "vars": ["V1"]
      },
      "x": {
        "val": [3, 3, 3, 3, 5, 3, 3, 1]
      }
    },
    config = {
      "barType": "lollipop",
      "colorScheme": "CanvasXpress",
      "dataPointSizeScaleFactor": 6,
      "graphType": "Bar",
      "widthFactor": 0.2,
      "sizeBy": "val",
      "xAxis": ["V1"]
    },
    width = 600,
    height = 600
  )
)
```
