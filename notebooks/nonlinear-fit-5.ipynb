# Unlocking Data Insights with Nonlinear Power Fit Graphs

Nonlinear Power Fit graphs provide insights into complex data relationships, essential for accurate modeling and prediction.

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/nonlinear-fit-5.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "nonlinearfit5",
    data = {
      "y": {
        "data": [
          [17.6, 159.9],
          [26, 206.9],
          [31.9, 236.8],
          [38.9, 269.9],
          [45.8, 300.6],
          [51.2, 323.6],
          [58.1, 351.7],
          [64.7, 377.6],
          [66.7, 384.1],
          [80.8, 437.2],
          [82.9, 444.7]
        ],
        "smps": ["Diameter", "Length"],
        "vars": ["S1", "S2", "S3", "S4", "S5", "S6", "S7", "S8", "S9", "S10", "S11"]
      }
    },
    config = {
      "decorations": {
        "power": [
          {
            "color": "red"
          }
        ]
      },
      "graphType": "Scatter2D",
      "showDecorations": True,
      "showLegend": False,
      "xAxis": ["Diameter"],
      "yAxis": ["Length"]
    },
    width = 600,
    height = 600
  )
)
```
