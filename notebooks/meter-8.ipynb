# Interactive Meter Graph for Metadata & Levels

Track metadata and levels in digital format with an interactive meter graph.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/meter-8.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "meter8",
    data = {
      "x": {
        "Quarter": ["Q1", "Q1", "Q1", "Q2", "Q2", "Q2", "Q3", "Q3", "Q3", "Q4", "Q4", "Q4"]
      },
      "y": {
        "data": [
          [1000, 5000, 4000, 2000, 7000, 9000, 8000, 3000, 2000, 6000, 5000, 10000]
        ],
        "smps": ["January", "February", "March", "April", "May", "June", "July", "August", "September", "October", "November", "December"],
        "vars": ["Revenue"]
      }
    },
    config = {
      "graphType": "Meter",
      "groupingFactors": ["Quarter"],
      "summaryType": "sum",
      "meterType": "digital",
      "meterVar": "Revenue",
      "xAxis": ["Revenue"]
    },
    width = 600,
    height = 600
  )
)
```
