# Multilevel Sankey with Equal Strata

The sankeyType 'equal' setting gives every stratum in a column the same height, evenly partitioning the column so rare categories stay readable.

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/sankey-8.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).
