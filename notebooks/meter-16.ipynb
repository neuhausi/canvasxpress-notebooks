# Vertical Meter Card

A vertical bar meter as a card with its own value axis; meterTitleAlign places the rotated title at the top (end).

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/meter-16.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "meter16",
    data = {
      "y": {
        "data": [
          [85]
        ],
        "smps": ["January"],
        "vars": ["Performance"]
      }
    },
    config = {
      "graphType": "Meter",
      "setMax": 100,
      "setMin": 0,
      "meterType": "vertical",
      "meterCard": True,
      "meterTitleAlign": "end",
      "xAxis": ["Performance"]
    },
    width = 600,
    height = 420
  )
)
```
