# 2D Function Scatter Plot

Visualize a 2D function scatter plot by altering the x-coordinate.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/functions-1.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "functions1",
    data = {
      "fx": "sin(x)"
    },
    config = {
      "graphType": "Scatter2D",
      "setMaxX": 24,
      "setMinX": 8
    },
    width = 600,
    height = 600
  )
)
```
