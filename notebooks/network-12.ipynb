# Unlock Data Insights with Network Graphs

Network graphs visually represent relationships between entities, enabling insights into centrality, clustering, and structures for data-driven decision-making.

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/network-12.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "network12",
    data = {
      "constraints": [
        {
          "axis": "x",
          "offsets": [
            {
              "nodeId": "Id1",
              "offset": "0"
            },
            {
              "nodeId": "Id2",
              "offset": "0"
            },
            {
              "nodeId": "Id3",
              "offset": "0"
            }
          ],
          "type": "alignment"
        },
        {
          "axis": "y",
          "offsets": [
            {
              "nodeId": "Id0",
              "offset": "0"
            },
            {
              "nodeId": "Id1",
              "offset": "0"
            },
            {
              "nodeId": "Id4",
              "offset": "0"
            }
          ],
          "type": "alignment"
        }
      ],
      "edges": [
        {
          "id1": "Id1",
          "id2": "Id2",
          "thickness": 3
        },
        {
          "id1": "Id2",
          "id2": "Id0",
          "thickness": 3
        },
        {
          "id1": "Id2",
          "id2": "Id3",
          "thickness": 3
        },
        {
          "id1": "Id2",
          "id2": "Id4",
          "thickness": 3
        }
      ],
      "nodes": [
        {
          "height": 20,
          "id": "Id0",
          "name": "a",
          "shape": "roundrect",
          "width": 30
        },
        {
          "height": 20,
          "id": "Id1",
          "name": "b",
          "shape": "roundrect",
          "width": 30
        },
        {
          "height": 20,
          "id": "Id2",
          "name": "c",
          "shape": "roundrect",
          "width": 30
        },
        {
          "height": 20,
          "id": "Id3",
          "name": "d",
          "shape": "roundrect",
          "width": 30
        },
        {
          "height": 20,
          "id": "Id4",
          "name": "e",
          "shape": "roundrect",
          "width": 30
        }
      ]
    },
    config = {
      "edgeColor": "rgb(122,78,79)",
      "edgeThickness": 3,
      "graphType": "Network",
      "networkColaAllConstraintsIterations": 10,
      "networkColaAvoidOverlaps": True,
      "networkColaHandleDisconnected": True,
      "networkColaLinkDistance": 80,
      "networkColaStartUnconstrainedIterations": 10,
      "networkColaUserConstraintIterations": 10,
      "networkLayoutType": "cola"
    },
    width = 600,
    height = 600
  )
)
```
