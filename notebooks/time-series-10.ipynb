# Dark-themed stacked areas with an italic footnote

Three size classes of catalogued near-Earth objects stacked as cumulative areas on a dark background, with a right-side legend and an italic footnote. Area graph with areaType stacked on a dark theme.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/time-series-10.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).
