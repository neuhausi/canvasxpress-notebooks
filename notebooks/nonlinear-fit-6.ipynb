# Polynomial Fit: Best Method for Nonlinear Regression

Polynomial fit: A versatile nonlinear regression method for modeling complex relationships in data analysis.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/nonlinear-fit-6.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "nonlinearfit6",
    data = {
      "y": {
        "data": [
          [-1, -1],
          [0, 3],
          [1, 2.5],
          [2, 5],
          [3, 4],
          [5, 2],
          [7, 5],
          [9, 4]
        ],
        "smps": ["X", "Y"],
        "vars": ["S1", "S2", "S3", "S4", "S5", "S6", "S7", "S8"]
      }
    },
    config = {
      "decorations": {
        "polynomial": [
          {
            "color": "red"
          }
        ]
      },
      "graphType": "Scatter2D",
      "showDecorations": True,
      "showLegend": False,
      "xAxis": ["X"],
      "yAxis": ["Y"]
    },
    width = 600,
    height = 600
  )
)
```
