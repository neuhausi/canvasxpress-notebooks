# Ring Meters with Corner Titles

Ring meters with meterRingTitlePosition set to topLeft place each tile's title in the card's top-left corner instead of centered above the circle — here the average satisfaction score per department.

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/meter-23.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "meter23",
    data = {
      "x": {
        "Department": ["Sales", "Sales", "Sales", "Sales", "Engineering", "Engineering", "Engineering", "Engineering", "Engineering", "Support", "Support", "Support"]
      },
      "y": {
        "vars": ["Satisfaction"],
        "smps": ["R1", "R2", "R3", "R4", "R5", "R6", "R7", "R8", "R9", "R10", "R11", "R12"],
        "data": [
          [78, 82, 80, 76, 91, 88, 93, 90, 89, 72, 70, 68]
        ]
      }
    },
    config = {
      "graphType": "Meter",
      "meterType": "ring",
      "groupingFactors": ["Department"],
      "summaryType": "average",
      "meterCard": True,
      "meterVar": "Satisfaction",
      "meterRingTitlePosition": "topLeft",
      "xAxis": ["Satisfaction"],
      "rangeColors": ["rgb(38,166,154)"]
    },
    width = 600,
    height = 600
  )
)
```
