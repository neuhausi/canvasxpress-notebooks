# Monthly climatogram: columns plus a line on two axes

Twelve months of rainfall as columns on the left axis and mean temperature as a line on the right axis, the classic dual-axis climate chart. BarLine graph with xAxis and xAxis2 series.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/time-series-17.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "timeseries17",
    data = {
      "y": {
        "vars": ["Rainfall (mm)", "Mean temperature (°C)"],
        "smps": ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"],
        "data": [
          [52.16, 41.42, 46.51, 49.26, 61.91, 62.53, 48.12, 60.43, 58.51, 65.3, 63.72, 60.81],
          [4.1, 5, 7.52, 9.49, 12.81, 16.13, 18.38, 18.15, 15.7, 11.45, 7.72, 5.19]
        ]
      }
    },
    config = {
      "graphType": "BarLine",
      "theme": "classic",
      "colorScheme": "User",
      "colors": ["#56B4E9", "#D55E00"],
      "title": "Wet winters, mild summers",
      "subtitle": "Monthly rainfall and mean air temperature at a synthetic temperate station",
      "graphOrientation": "vertical",
      "xAxis": ["Rainfall (mm)"],
      "xAxis2": ["Mean temperature (°C)"],
      "xAxis2Show": True,
      "xAxisTitle": "Rainfall (mm)",
      "xAxis2Title": "Temperature (°C)",
      "lineThickness": 3,
      "lineType": "spline",
      "lineDecoration": "symbol",
      "dataPointSize": 22,
      "legendPosition": "top",
      "smpTitle": "Month",
      "smpTextRotate": 90
    },
    width = 760,
    height = 520
  )
)
```
