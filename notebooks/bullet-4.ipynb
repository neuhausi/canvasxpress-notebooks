# Visualization of Bullet Graphs

Clearly visualize performance with bullet graphs, incorporating adjusted scales, target markers, and qualitative ranges.

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/bullet-4.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "bullet4",
    data = {
      "y": {
        "data": [
          [230, 180, 185, 225, 80, 190],
          [100, 15, 35, 115, 95, 30],
          [150, 50, 70, 170, 135, 50],
          [250, 250, 250, 250, 250, 250],
          [234, 120, 150, 230, 175, 55]
        ],
        "smps": ["USA", "Chile", "Brazil", "Canada", "Mexico", "Argentina"],
        "vars": ["Value", "Low", "Average", "High", "Target"]
      }
    },
    config = {
      "colors": ["#305DCB", "#E4EFFD", "#ADCFFA", "#73AFF8"],
      "dataValuesPosition": "inside",
      "dataTextColor": "#FFFFFF",
      "graphOrientation": "horizontal",
      "graphType": "Bullet",
      "showDataValues": True,
      "xAxis": ["Value"],
      "rangeStack": ["Low", "Average", "High"],
      "rangeColors": ["#73AFF8", "#ADCFFA", "#E4EFFD"],
      "bulletTargetVarName": "Target"
    },
    width = 600,
    height = 600
  )
)
```
