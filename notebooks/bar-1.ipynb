# Simple Bar Graph for Easy Comparison

A basic bar graph displaying a single series for easy comparison.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/bar-1.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "bar1",
    data = {
      "y": {
        "data": [
          [4, 5, 4, 4, 7]
        ],
        "smps": ["Cat 1", "Cat 2", "Cat 3", "Cat 4", "Cat 5"],
        "vars": ["Var 1"]
      }
    },
    config = {
      "graphOrientation": "vertical",
      "graphType": "Bar",
      "showLegend": False,
      "smpTextRotate": 90,
      "smpTitle": "Categories",
      "title": "Bar graph with a single series",
      "xAxis": ["Var 1"],
      "xAxisTitle": "Var 1"
    },
    width = 600,
    height = 600
  )
)
```
