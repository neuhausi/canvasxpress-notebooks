# Custom bracket colors with BY correction

Wilcoxon tests on mtcars fuel economy by cylinder count, Benjamini-Yekutieli adjusted. Each bracket has its own color, and the original data points are overlaid.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/statistics-7.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "statistics7",
    data = {
      "y": {
        "vars": ["mpg"],
        "smps": ["Obs1", "Obs2", "Obs3", "Obs4", "Obs5", "Obs6", "Obs7", "Obs8", "Obs9", "Obs10", "Obs11", "Obs12", "Obs13", "Obs14", "Obs15", "Obs16", "Obs17", "Obs18", "Obs19", "Obs20", "Obs21", "Obs22", "Obs23", "Obs24", "Obs25", "Obs26", "Obs27", "Obs28", "Obs29", "Obs30", "Obs31", "Obs32"],
        "data": [
          [22.8, 24.4, 22.8, 32.4, 30.4, 33.9, 21.5, 27.3, 26, 30.4, 21.4, 21, 21, 21.4, 18.1, 19.2, 17.8, 19.7, 18.7, 14.3, 16.4, 17.3, 15.2, 10.4, 10.4, 14.7, 15.5, 15.2, 13.3, 19.2, 15.8, 15]
        ]
      },
      "x": {
        "cyl": ["4", "4", "4", "4", "4", "4", "4", "4", "4", "4", "4", "6", "6", "6", "6", "6", "6", "6", "8", "8", "8", "8", "8", "8", "8", "8", "8", "8", "8", "8", "8", "8"]
      }
    },
    config = {
      "graphType": "Boxplot",
      "graphOrientation": "vertical",
      "groupingFactors": ["cyl"],
      "xAxis": ["mpg"],
      "xAxisTitle": "mpg",
      "smpTitle": "cyl",
      "smpTitleFontStyle": "bold",
      "axisTitleFontStyle": "bold",
      "axisAlgorithm": "rPretty",
      "smpTextRotate": 0,
      "showLegend": False,
      "colorBy": "cyl",
      "title": "Fuel economy by cylinders (Wilcoxon, BY-adjusted)",
      "showStatInference": True,
      "statInference": {
        "correction": {
          "method": "BY"
        },
        "annotate": "padj",
        "comparisons": [
          {
            "groups": ["4", "6"],
            "p": 0.000665814784145945,
            "color": "#1b9e77"
          },
          {
            "groups": ["6", "8"],
            "p": 0.00101304468807941,
            "color": "#d95f02"
          },
          {
            "groups": ["4", "8"],
            "p": 2.77471487941046e-05,
            "color": "#7570b3"
          }
        ]
      },
      "showBoxplotOriginalData": True,
      "jitter": True,
      "colorScheme": "Dark2"
    },
    width = 600,
    height = 600
  )
)
```
