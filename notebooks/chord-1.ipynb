# Visualize Relationships with Chord Graphs

Chord graphs offer a clear visualization of relationships in complex datasets, making intricate connections easily understandable in a circular layout.

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/chord-1.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "chord1",
    data = {
      "y": {
        "data": [
          [11975, 5871, 8916, 2868],
          [1951, 10048, 2060, 6171],
          [8010, 16145, 8090, 8045],
          [1013, 990, 940, 6907]
        ],
        "smps": ["A", "B", "C", "D"],
        "vars": ["A", "B", "C", "D"]
      }
    },
    config = {
      "circularArc": 360,
      "circularRotate": 0,
      "circularType": "chord",
      "colors": ["#000000", "#FFDD89", "#957244", "#F26223"],
      "graphType": "Circular",
      "higlightGreyOut": True,
      "legendKeyBackgroundBorderColor": "rgba(255,255,255,0)",
      "legendKeyBackgroundColor": "rgba(255,255,255,0)",
      "objectBorderColor": "rgb(0,0,0)",
      "rAxisTickFormat": ["%sK", " / 1000"],
      "showTransition": False,
      "title": "Simple Chord Graph",
      "transitionStep": 50,
      "transitionTime": 1500
    },
    width = 600,
    height = 600
  )
)
```
