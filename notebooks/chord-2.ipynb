# Rotated Chord Graph Reveals Data Relationships

Chord graph displayed in a rotated layout, showcasing intricate relationships between data points.

Live example, data and code in other languages: https://www.canvasxpress.org/examples/chord-2.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).

```python
from canvasxpress.canvas import CanvasXpress
from canvasxpress.plot import graph

graph(
  CanvasXpress(
    render_to = "chord2",
    data = {
      "y": {
        "data": [
          [11975, 5871, 8916, 2868],
          [1951, 10048, 2060, 6171],
          [8010, 16145, 8090, 8045],
          [1013, 990, 940, 6907]
        ],
        "smps": ["A", "B", "C", "D"],
        "vars": ["A", "B", "C", "D"]
      }
    },
    config = {
      "chordColor": "largest",
      "circularArc": 360,
      "circularRotate": 180,
      "circularType": "chord",
      "graphType": "Circular",
      "higlightGreyOut": True,
      "legendKeyBackgroundBorderColor": "rgba(255,255,255,0)",
      "legendKeyBackgroundColor": "rgba(255,255,255,0)",
      "objectBorderColor": "rgb(0,0,0)",
      "rAxisTickFormat": ["%sK", " / 1000"],
      "showTransition": False,
      "title": "Rotated Chord Graph",
      "transitionStep": 50,
      "transitionTime": 1500
    },
    width = 600,
    height = 600
  )
)
```
