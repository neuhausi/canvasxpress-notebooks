# Master small datasets with dot plots

Dot plots are a useful tool for visually displaying small datasets, allowing for easy identification of clusters and outliers in the data.

Live example, data and R/JavaScript code: https://www.canvasxpress.org/examples/dotplot-13.html

Requires `pip install "canvasxpress[jupyter]"` (preinstalled on Binder).
